## Abstract

Build and evaluate a small reference library for expressive local speech, starting from the reference-guided continuation method that the user found effective in Paper 15. Extend beyond the five available broad emotion categories while testing intelligibility, believable delivery and speaker continuity. This notebook begins as a study design; no new performance references or validation results are claimed.

## Starting evidence

Paper 15’s final neutral-versus-angry reference comparison used the same speaker, target words, transcript metadata, seed and native continuation mode with no written emotion direction. The user explicitly reported that it worked and conveyed emotion. That is the positive starting result for Paper 16, rather than a claim that the earlier global instruction-switching tests succeeded.

## Research question

Can a small set of carefully selected, same-speaker reference performances reliably guide subtler deliveries across new text while preserving the intended speaker and every word?

## Hypotheses and alternatives

H1: selected references produce recognizable intended deliveries on new lines. H2: speaker identity and wording survive those changes. H3: references generalize across more than the line used to select them. Alternatives include references transferring mostly pitch/timbre, content-dependent effects, unstable identity, weak contrast and cherry-picked takes.

## Initial range and scope

Start with one existing speaker and preserve the five-category baseline. Candidate additions: warm reassurance, curiosity, skepticism, disappointment and restrained frustration; intensity is a separate dimension to test, not merely another emotion label. These are proposed targets, not existing validated reference files. Establish a small reliable set before expanding to more speakers.

## Reference acquisition and provenance

Reuse existing eligible recordings first. Any new reference must record origin, terms/permission status, speaker identity basis, exact transcript, acoustic preparation, duration and hash. Synthetic candidates must be labeled synthetic and auditioned before promotion. An emotion name or generation prompt does not establish that a clip performs it. Public availability alone is not recorded as permission to reuse.

## Experiment design

Hold target text, model revision, voice identity anchor, continuation mode, seed and sampling settings fixed when comparing references. Verify native mode mapping and preserve raw audio. Select references on development lines, then test on fresh lines not used to choose them. Record repetitions/seeds and all failures. Avoid presenting cached or selected favorites as independent evidence.

## Implementation and artifacts

Notebook-controlled generation; per-reference manifest; per-line requests and native output files; static-gain listening copies; source/model hashes; timing and device evidence; listener judgments and keep/retake reasons. No new interface, production-default replacement, download or model installation is required by this initial template.

## Listening and technical evaluation

Technical checks: finite non-silent readable output, source/request integrity and playback-level matching. Listener checks: intended delivery, naturalness, exact words, identity continuity, artifacts and usefulness. Acoustic pitch/duration measurements remain diagnostics. Keep scores empty until assessed. Describe who listened and whether judgments were blinded.

## Acceptance and failure rules

A reference is only a candidate until tested on multiple new lines and repeated generations. Predefine promotion thresholds before validation; record disagreement rather than forcing a winner. Missing words, unstable identity, inappropriate emotion or inaudible contrast require rejection or retake. Signal differences alone do not qualify a reference.
