## 6. Artifact schemas and provenance
Artifact definitions use JSON Schema 2020-12 with required fields, closed object properties, explicit arrays/enums, timestamps and conditional requirements. A retrieved source requires a snapshot reference and hash; failed retrieval retains a reason. Supported judgments require evidence, an entailment review and reviewer reference. Acknowledgment requires an acknowledgment record.

These constraints establish record shape, not truth. A validator can require a source reference, but only an artifact registry and inspection can show that the source exists and supports a claim. Conditional 'accepted review' schema checks are supplemented by the complete critical-review rubric check.

| Record group | Purpose | Additional semantic obligation |
|---|---|---|
| Brief/plan/discovery | Scope, query intent and candidate screening | Allowed source/access and bounded search coverage |
| Source/evidence | Actual source capture, observation context, transformations and dependence | Resolve every reference and preserve raw evidence |
| Alternatives/candidate claims/assessments | Separate proposals from judgments | Entailment, scope and contradictory evidence review |
| Dossier/review/receipt | Decision-ready report and attributable acceptance | Exact revision match, reviewer authority and acknowledgment |
| Costs/obligations/change/impact/revision | Honest limits, accounting and controlled change | Real events, no duplicated costs, approved immutable revisions |

The record contracts retain unknown amounts, uncertain independence and empty evidence without inventing values. No source URL is treated as an immutable snapshot. A separate activation schema rejects unresolved or wrongly typed limits and requires real role/runtime bindings; passing this schema still cannot prove that those bindings are authorized or functional. External metrics require source-preserved units/period/context; extending metric-specific schema is required before accepting analytics as a direct connector input.

````python
schemas=load_schema()
from jsonschema import Draft202012Validator
Draft202012Validator.check_schema(schemas)
display(pd.DataFrame([{"Artifact type":k,"Required payload fields":", ".join(v["properties"]["payload"]["required"])} for k,v in schemas["$defs"].items()]))
print("Artifact schema grammar valid. Cross-artifact facts and source truth remain separate checks.")
````

## 8. Lifecycle, authority, failure and recovery
The JSON specifies states, events, actor roles, guard references and cross-cutting exception routes. Guard context is Boolean evidence supplied by named systems: configuration/authority/registry/budget/review records. Missing values and text such as 'true' do not grant permission. The local controller enforces type/reference checks, deadline/cancellation, checkpoints, reviewer identity/revision checks, and stage leases. Local trusted callers supply review decisions; this is not an authentication boundary against code running as the same OS user.

| Situation | Required outcome |
|---|---|
| Missing scope | Clarification obligation; no collection |
| Limit/access failure | Record attempt/cost and block; no authorization bypass |
| Retrieval unavailable | Retain failure; allow qualified insufficient output |
| Review fails | Reject or request a child revision; preserve submission |
| Delivery fails | Preserve accepted dossier; bounded duplicate-safe retry |
| New evidence | Impact report, approved new brief and immutable child revision |
| Cancellation | Propagate to leased work, record partial costs and artifacts |

The dossier carries a cost snapshot through assembly. Review and delivery add more cost, so the final run ledger must be linked separately at closure. Do not silently mutate the accepted dossier or pretend its earlier snapshot includes later work. SQLite implements immutable artifacts, compare-and-swap run updates, append-only events, stage leases and a duplicate-safe local outbox. Review waiting pauses the remaining active budget without replenishing it. Structured reviewer feedback creates a child run with reviewed assessments and a reasoned continue/narrow/ask/stop/insufficient recommendation; a continue decision requires at least one supported reviewed claim. Parent artifacts are never overwritten. Cancellation blocks subsequent stages; an in-flight blocking call ends within its timeout. Consumer notifications are local outbox records, not a claim that another instance is already deployed.

### 14.7 Semantic audit, current evidence and a usable research decision

**The typed-guided result is not accepted as useful intelligence.** Its citations resolve, but “private educational-content pilot” is not a specific concept; its counterargument merely compares issue counts; and “CLI works normally” does not measure an educational benefit. Section 14.6's promotion was a traceability-only development output, superseded by this audit. This is precisely why a schema pass must not be called finished intelligence.

The final decision procedure below uses explicit, inspectable rules rather than hiding weak model prose: screen **recently created**, open, title-relevant Mac problems; group by a named problem family; require at least two different reporting authors; retain official documentation as an existing alternative; and produce a concrete diagnostic-content research brief. This is authored decision logic, not mined evidence that the market wants videos. A software defect may require an upstream fix, so the content concept is **diagnosis and supported-path selection**, not a promised bug fix.

A 180-day window, two-author minimum, and two private comparison tasks are declared study policies, not empirically optimal thresholds. The current outputs can justify a private feasibility experiment. They cannot establish channel demand, willingness to pay or profit.


````python
# Final bounded decision mechanism. No owner review is needed to select a research priority.
cutoff=datetime.datetime.fromisoformat(CAPTURED_AT)-datetime.timedelta(days=180)
FAMILIES={
    "memory_and_backend":{"pattern":r"\bMPS\b|\bMetal\b|\bMLX\b|\bOOM\b|memory|allocation|panic|backend|performance", "question":"How can a Mac user diagnose local-AI memory/backend failures and choose a documented supported next step?"},
    "installation":{"pattern":r"install|download|binaries|compil|app window", "question":"How can a Mac user distinguish a supported installation path from a missing or incompatible build?"}}
GROUPS=[];EXCLUSIONS=[]
for topic in TOPICS:
    recent=[]
    for s in ISSUES:
        if s['topic']!=topic:continue
        created=datetime.datetime.fromisoformat(s['created_at'].replace('Z','+00:00'))
        title_mac=bool(MAC.search(s['title'])) or bool(re.search(r'\bmlx\b',s['title'],re.I))
        reason=None
        if s['state']!='open':reason='closed: retain as opposing/context evidence'
        elif created<cutoff:reason='created outside the 180-day window; updated time is not new demand'
        elif not title_mac:reason='Mac relevance not explicit in title; exclude conservatively from final ranking'
        if reason:EXCLUSIONS.append({'source_id':s['id'],'reason':reason})
        else:recent.append(s)
    for family,cfg in FAMILIES.items():
        members=[s for s in recent if re.search(cfg['pattern'],s['title'],re.I)]
        GROUPS.append({'candidate':topic,'family':family,'issues':len(members),'authors':len({s['author'] for s in members}),'source_ids':[s['id'] for s in members],'eligible':len({s['author'] for s in members})>=2 and topic+'-official-docs' in INDEX})
GROUP_TABLE=pd.DataFrame(GROUPS).sort_values(['eligible','authors','issues','candidate','family'],ascending=[False,False,False,True,True]).reset_index(drop=True)
display(GROUP_TABLE);display(pd.DataFrame(EXCLUSIONS))
chosen=next((g for g in GROUP_TABLE.to_dict('records') if g['eligible']),None)
if chosen:
    topic=chosen['candidate'];family=chosen['family'];sources=[INDEX[sid] for sid in chosen['source_ids']]
    FINAL_DECISION={
      'status':'research-priority-identified; market-viability-not-established',
      'audience':'Mac users attempting local AI, initially the developer issue-reporting segment',
      'candidate':topic,'problem_family':family,
      'content_concept':f"{topic} on Mac: a version-aware diagnostic decision tree for {family.replace('_',' ')} failures",
      'problem_question':FAMILIES[family]['question'],
      'observed_support':{'distinct_reports':len(sources),'distinct_reporting_authors':len({s['author'] for s in sources}),'repository_families':1,'window_days':180},
      'evidence':[{'source_id':s['id'],'reported_problem':s['title'],'url':s['url'],'created_at':s['created_at'],'state':s['state'],'reproduced':False} for s in sources],
      'existing_alternative':{'source_id':topic+'-official-docs','url':TOPICS[topic]['docs'],'coverage_gap':'not established'},
      'opposing_case':'Official documentation may already answer the questions. Reported crashes can be version-specific upstream defects that an educational video cannot fix. GitHub users may prefer written troubleshooting to video.',
      'uncertainty':'One exposed repository family, convenience sampling and title-based eligibility; no measured viewing intent, comprehensive competitor survey, verified remedy, audience trial or revenue.',
      'next_action':'Build a private diagnostic prototype only; do not present an unverified remedy as a fix.',
      'private_pilot':{
        'tasks':['Given a version/hardware/error report, find the matching known issue and supported documentation.', 'Given an unsupported or unverified configuration, explain the uncertainty and choose a safe next investigation rather than inventing a repair.'],
        'comparison':'Use the same two report scenarios with official documentation alone versus the original diagnostic prototype. Hold task inputs constant; record correctness, unsupported advice and elapsed time.',
        'success_rule':'Both tasks must be correctly resolved or explicitly identified as unresolved, with zero unsupported repair claims. Prototype must improve at least one prespecified task measure without worsening correctness; a tie does not establish benefit.',
        'stop_rule':'Stop/narrow if original diagnostics add no measurable benefit, the reports cannot be interpreted safely, a remedy cannot be verified, or the declared production budget is exceeded.',
        'observations':None,'executed':False},
      'operating_costs':{'measured_model_seconds':round(sum(c['seconds'] for c in MODEL_AUDIT)+sum(c['seconds'] for c in PAIRED),3),'measured_acquisition_seconds':round(sum(a.get('seconds',0) for a in AUDIT),3),'money_spent_on_remote_paid_services':0,'electricity_cost':None,'human_time_cost':None,'production_cost':None},
      'commercial_decision':'Do not scale, spend or claim market viability from this study.',
      'profitability_proven':False}
else:FINAL_DECISION={'status':'insufficient','reason':'No recent title-relevant problem family has two distinct reporting authors and a retrieved official alternative.'}
display(JSON(FINAL_DECISION,expanded=True))
# A practical, source-linked brief rather than an opaque graph.
if chosen:
    display(Markdown('#### Research decision: '+FINAL_DECISION['content_concept']))
    display(Markdown('**Why investigate:** '+str(FINAL_DECISION['observed_support']['distinct_reports'])+' recent sampled reports by '+str(FINAL_DECISION['observed_support']['distinct_reporting_authors'])+' reporting authors describe this problem family. This is a problem signal, not audience demand.'))
    for item in FINAL_DECISION['evidence']:display(Markdown(f"- [{item['reported_problem']}]({item['url']}) — reported, not reproduced"))
    display(Markdown('**Existing alternative:** [official documentation]('+FINAL_DECISION['existing_alternative']['url']+'). No coverage gap established.'))
    display(Markdown('**What could make us reject it:** '+FINAL_DECISION['opposing_case']))
# Preserve an explicit distinction between traceability acceptance and useful recommendation acceptance.
SEMANTIC_REVIEW=pd.DataFrame([
 {'version':'free-form direct','citation_gate':DIRECT_AUDIT['passed'],'usefulness_accepted':False,'reason':'Wrong competitor, invalid decisions and unverified engagement target'},
 {'version':'free-form staged','citation_gate':STAGED_AUDIT['passed'],'usefulness_accepted':False,'reason':'Drifted into software release rather than educational content'},
 {'version':'typed-guided','citation_gate':next(t['audit']['passed'] for t in PAIRED if t['arm']=='typed-guided'),'usefulness_accepted':False,'reason':'Generic concept and weak counterargument; traceability alone was insufficient'},
 {'version':'explicit final research procedure','citation_gate':bool(chosen),'usefulness_accepted':bool(chosen),'reason':'Accepted only as a specific private research brief, not a validated market opportunity'}])
display(SEMANTIC_REVIEW)
````
