## Whole-business design: seven specialized instances and shared planes
**Research date: 30 October 2026 · Provisional business plan · No content published**

### Abstract
We investigate whether an original social-media content business can operate with locally generated scripts, narration and visual assets, minimal manual intervention, and no calls or meetings, while covering both cash expenditure and the owner's time. Primary-source policy and model-license research is combined with a read-only local hardware inventory, alternative monetization designs and explicitly hypothetical economic sensitivity analysis. We propose an initial practical software-explainer format, using original demonstrations and diagrams rather than relying on expensive fully generated video. The design proceeds from a business goal to seven specialized instances—Market Intelligence, Content Strategy, Production, Distribution, Monetization, Audience Care, and Finance and Performance—coordinated by Operations. Each instance shares the same Configuration, Components, Composites, Flows and Processes planes while retaining its own responsibilities and state. Explicit artifact contracts and feedback across revisions connect the instances. Measured local production and audience evidence are required before scale. Advertising, affiliate and self-service product revenue remain conditional alternatives. This paper establishes a research-backed plan and reproducible scenario arithmetic; it establishes neither market demand, production throughput, monetization acceptance nor profitability.

**Architecture revision:** the earlier production-centered single-instance sketch is preserved in the study revisions folder and superseded by Section 6’s whole-business design. This revision adds no production deployment or commercial evidence.

**Five phases:** establish constraints and falsifiable hypotheses → curate demand and rights evidence → produce a bounded private local pilot → authorize and measure a limited audience experiment separately → retain, revise or stop based on observed value and economic cost. Only the research/design phase and arithmetic are completed here. Existing papers and the active canvas are preserved.

## 1. Research question and scope
**Question:** Can a locally produced, original educational-content operation deliver useful work to a specific audience with sufficiently low recurring human effort to become economically sustainable, without calls or meetings?

| Constraint | Working interpretation | What is not assumed |
|---|---|---|
| Social media | YouTube is the initial research reference; secondary distribution is deferred | A measured best platform or immediate ad access |
| Local AI content | Local script/scene generation and synthetic narration; original locally rendered visuals | All pixels generated by a model; offline research/distribution |
| Maximum automation | Automate repeatable production and bookkeeping; expose review and exception queues | Zero human judgment, customer messages or verification |
| Revenue North Star | Track receipts, costs, economic profit and cash runway | Views, upload count or estimated platform revenue equal cash |
| Iterative instance | Explicit provisional requirements and revision history | Complete one-shot generation or automatically discovered hierarchy |
| Current authorization | Research, plan, scenario mathematics and notebook delivery | Publishing, buying, opening accounts or deploying production |

**Stop condition for this phase:** a traceable conditional recommendation, testable assumptions, cost model and layered design—not a revenue claim.

## 2. Critical thinking: hypotheses and alternatives
| ID | Hypothesis | Evidence that would support it | What would challenge it |
|---|---|---|---|
| H1 | A narrow practical problem supports materially different useful episodes | Independent audience/problem signals and later observed viewer response | Only broad trends, copied ideas or generic AI summaries |
| H2 | Local generation can meet the quality/time envelope | Reproducible private end-to-end artifacts with measured operator minutes and correctness | Excess correction, inaccessible environment or unclear rights |
| H3 | A viable revenue path can cover economic cost | Actual eligible receipts exceed costs under a bounded pilot | Works only by omitting owner time, assuming immediate YPP or optimistic rates |
| H4 | A layered representation improves inspectability and revision | Each output traces to requirements, inputs, definitions and checks; unrelated edits stable | Merely renamed boxes, no enforceable handoffs or traceable changes |

**Competing explanations:** interest may reflect tool novelty rather than durable need; good output may come from manual editing rather than automation; view gains may come from topic selection rather than the production architecture. Keep these explanations alive. Automation and business viability are separate hypotheses. No success is predetermined. The private pilot should also compare the automated workflow with a manual or partially automated reference, so that reduced owner effort is measured rather than assumed.

## 3. Business recommendation and revenue mechanism
Start with a **conditional original practical software-explainer pilot**, preferably a narrow class of repeatable tasks the owner can independently verify. Example directions are local AI workflows, spreadsheet automation or developer productivity; these are candidate niches, not measured demand winners. Produce one useful result and explain it with evidence, not generic news narration.

| Path | Who pays and why | Key gate | Operational burden | Decision |
|---|---|---|---|---|
| Ads/Premium | Platform shares eligible viewing revenue | Audience thresholds, review, eligible account and viewing | Low buyer fulfillment; editorial/account work remains | Later revenue option; zero assumed before eligibility |
| Relevant affiliate links | Merchant pays for attributed purchases | Program acceptance, truthful demonstrated usefulness and purchase | Link updates, disclosure and exceptions | Optional only after product knowledge and terms review |
| Self-service artifacts | Viewer buys a tested useful file/tool | Real buyer need, eligible checkout and trustworthy delivery | Product maintenance, refunds and written support | Optional experiment; not needed for first media proof |
| Sponsorship | Brand pays for negotiated exposure | Negotiation, approvals and audience | Higher contact burden | Excluded from initial model |

Do not choose ads-only merely because fulfillment is easy. Do not build a product merely to avoid ad thresholds. First establish a specific audience problem, then compare the acceptable burden and evidence for each payment path. A checkout is not demand.

**Value proposition (provisional):** concise, tested demonstrations that save a narrowly defined audience time, with reproducible steps and clearly disclosed limitations. **Moat hypothesis:** reliable original demonstrations and a structured evidence library; generation volume is not a moat.

### Demand research protocol before production commitment
Define two narrow problem/audience pairs. For each, collect a reproducible dated sample of primary questions/search-interest evidence and existing treatments, preserve query/date/source and deduplicate copied material. Distinguish search interest, competitor views and willingness to pay. Score specificity, demonstration feasibility, differentiation, repeatability and contact burden; publish raw observations alongside judgments. Public-source research can be done asynchronously. Do not infer addressable revenue from keyword counts.

| Record | Required fields |
|---|---|
| Demand observation | ID, query, date, source URL, audience, problem, raw observation, limits, duplicate family |
| Claim | ID, statement, source IDs, fact/assumption/target classification, freshness, reviewer decision |
| Candidate niche | audience, repeated problem, original treatment, evidence IDs, rights risks, owner expertise, unresolved gaps |
| Decision | version, alternatives, reason, evidence, counterevidence, revisit trigger |

**Selection gate:** no niche is selected until we can state one useful viewer outcome and show independent evidence for the problem. This paper does not fabricate that evidence.

## 5. Pipeline and local implementation candidates

The pipeline produces a verified **release package**, not automatic public publication. No runtime is installed or started by this plan.

## 6. Whole-business architecture: seven instances, shared planes
The business is represented as a **system of specialized instances**. Each instance has the same five internal planes: Configuration, Components, Composites, Flows and Processes. The business itself is an enclosing scope, not a renamed content-production instance. **Operations** coordinates these instances; it is not asserted as an additional universal plane or a separate eighth specialist. This domain model is an authored proposal to test, not a discovered natural hierarchy.

The distinction is responsibility: Market Intelligence discovers needs; Content Strategy chooses what to address; Production makes artifacts; Distribution manages channel/release state; Monetization manages revenue routes; Audience Care handles trust and service; Finance and Performance accounts for costs and results. They share evidence and goals without sharing unrestricted control.

| Business lifecycle | Responsible instance | Evidence / outgoing result | Important boundary |
|---|---|---|---|
| Discover a problem | Market Intelligence | Dated opportunity dossier, contrary evidence and uncertainty | Popularity is not demand or payment |
| Decide the promise | Content Strategy | Prioritized viewer outcome and acceptance rubric | Commercial pressure cannot replace editorial usefulness |
| Make the artifact | Production | Private release package, provenance and total attempt cost | Rendering does not prove value |
| Reach the audience | Distribution | Authorized release receipt and metric snapshots | Publication disabled in this study |
| Capture value | Monetization | Eligible route, relevant offer and attributable receipt claim | No eligibility or revenue assumed |
| Maintain trust | Audience Care | Feedback, corrections, optional product support and resolutions | Written asynchronous contact can remain |
| Reconcile and learn | Finance and Performance | Actual costs, reconciled receipts and decision reports | Estimated revenue, payable balance and received cash stay separate |
| Coordinate the whole | Operations | Priorities, caps, approvals and versioned decisions | Coordinates specialists; does not bypass their checks |

**Shared invariants:** original useful content; no calls/meetings by design; local media generation; lawful source/asset use; real event records; explicit unknowns; no automatic publication, purchase or account changes. Online research, distribution and payments are not claimed to run locally. Native Langflow definitions remain unchanged; external local programs need real adapters or resource references.

The complete business has feedback loops. It must **not** be forced into one global DAG. A particular artifact's dependency/revision history can be acyclic; performance from one episode informs a *later* brief. Cyclic business learning is represented by time and version, not a same-version causal loop. No causal improvement is established by drawing a feedback arrow.

Top-down: intent → domain responsibilities → processes → flows → bounded capability requirements. Bottom-up: actual configuration/resources → verified native component occurrences → validated composites → coordinated flows → accountable processes. Middle-out revisions remain allowed. Unsupported requirements stay visible rather than being hidden to complete the tower.

## 7. Economics: uncertainty before forecasts
Let \(V\) be monthly total views and \(r\) an **assumed** net ad receipt per 1,000 total views. With eligibility indicator \(e\), illustrative receipts are \(R=eVr/1000\). This is not advertiser CPM; the input must eventually be calibrated to the channel and viewing mix. Before eligibility, \(e=0\).

For attempts \(N\), accepted fraction \(a\), generation hours \(h\), assumed mean power \(p\), energy price \(q\), review minutes \(m\), other operator hours \(u\), owner hour value \(w\), fixed cash \(F\) and hardware economic charge \(H\):

\[C_{cash}=F+Nhpq,\quad C_{economic}=C_{cash}+(Nm/60+u)w+H.\]
\[\Pi_{economic}=R-C_{economic},\qquad V_{break-even}=1000C_{economic}/r\quad(e=1,r>0).\]
Expected accepted output \(Na\) is scenario arithmetic, not a throughput result. Do not derive views from video count. Rejected attempts still cost time. An all-failed period retains costs and reports unit cost as undefined. The compact model is not a full tax, payment-delay or runway ledger. Hardware charge is an economic allowance, separate from incremental cash; avoid double counting a new purchase and its full amortization.

| Input | Must ultimately come from | Current status |
|---|---|---|
| Views and net receipts | Actual eligible platform/account observations | Unknown; scenarios only |
| Runtime, review, acceptance | Recorded private pilot attempts, including failures | Unmeasured |
| Power and energy price | Meter/credible local tariff | Unmeasured/unknown |
| Owner time and hardware charge | Explicit owner valuation and accounting choice | Authored assumptions |
| Store/affiliate economics | Actual program fees, returns, conversion and support | Excluded from ad arithmetic; separate model required |

**All numbers below are arbitrary USD scenarios, not market estimates, measured laptop power, expected audience or an earnings forecast.** No probability distributions are fitted, so there is no Monte Carlo precision theater.

## Low-cost validation before any publication

This protocol is proposed, not executed. Its purpose is to reject weak options cheaply; it cannot prove a profitable business privately.

1. **Freeze the decision frame.** Choose two candidate audience/job combinations, one language, a target viewing region, a research time cap, and a maximum owner review time per artifact. Record undecided inputs rather than silently selecting English or a country.
2. **Predeclare a comparable observation sample.** For each candidate, define the same number of task queries, time window, result-depth limit, inclusion criteria, and original-source deduplication rule. Choose the budget before seeing results. Save query, access date, source URL, result position, tool/content version, apparent user job, and the concrete gap being claimed. A copied video or syndicated page is not independent corroboration.
3. **Triangulate.** Compare relative Trends interest, available platform content-gap signals, and manual inspection of current tutorials. Record good existing solutions and reasons the proposed content would add nothing. A missing search result is not automatically an opportunity: it can indicate missing demand.
4. **Define one differentiated promise.** For example, a reproducible local workflow with explicit hardware requirements and failure handling. Specify exactly how the viewer verifies success. Reject broad “AI news” or “make money with AI” positioning without a defensible audience advantage.
5. **Design offline production trials.** Parent implementation may later produce matched pilot artifacts and log compute, failures, owner minutes, corrections, intelligibility, accessibility, rights, and reproducibility. This research task does not perform those trials. Internal reviewers can judge correctness and clarity; they cannot substitute for audience demand.
6. **Model uncertainty before release.** Include no-income startup, low-conversion, refund, update and platform-unavailability cases. Leave unobserved revenue parameters symbolic or explicitly assumed. Do not present a break-even calculation as a forecast.
7. **Make the next gate explicit.** Only an authorized audience-facing experiment can establish actual watching or buying. At that later gate, agree the budget and metrics before publication; keep early results separate from the data used to revise the content.

Use a common accounting period and avoid double counting. Owner labor is a cost even when no cash wage is paid.

## Critical review

The conditional pilot is defensible as a **research hypothesis**, not yet a business. Original practical software explainers offer something automation alone cannot: demonstrations whose usefulness and correctness can be checked. The central tension is that **maximum automation may reduce the expertise, differentiation and trust needed to earn revenue**.

### Strongest disconfirming risks

1. **No distinctive audience value.** Accurate explanations may still be interchangeable with documentation, search results and established creators. Faster production does not establish a reason to watch.
2. **Automation shifts rather than eliminates labor.** Technical verification, demonstration maintenance, licensing, editorial judgment and exceptions may dominate owner time. A pipeline that needs frequent corrections could defeat the intended operating model.
3. **Distribution and monetization are separate bottlenecks.** Upload frequency does not imply views; views do not imply retention, eligibility or earnings. Originality and platform review remain relevant even with genuinely original assets.
4. **Local feasibility may not translate into acceptable economics.** M3 Max/64 GiB is a resource constraint, not evidence of throughput, narration quality or unattended reliability. Include rerenders, idle capacity and failure recovery—not just successful generation.
5. **“No meetings” does not mean “no support.”** Affiliate recommendations create disclosure and trust obligations; self-serve products create maintenance, refunds and asynchronous support. These routes could introduce more operational burden than ads.
6. **Dependency and factual risk.** Model and voice licenses, software demonstration rights, privacy, changing interfaces and misleading technical claims require review. Excluding noncommercial F5 weights is appropriate; it does not clear the remaining stack.

A particularly damaging outcome would be: technically polished videos that require substantial expert correction and still have no identifiable advantage for a specific viewer.

### What must be decided versus measured

**Decide before pilot design**
- Actual residence, payment and relevant platform eligibility.
- One audience, language, narrow problem domain and credible expertise boundary.
- Available owner hours, cash ceiling and acceptable hardware economic charge.
- Whether monetization priority is ads, affiliate referrals or a product—each implies different content and obligations.
- Acceptable asynchronous support burden and prohibited claims/actions.

**Measure in bounded private research**
- Factual correctness and reproducibility of demonstrations.
- Active owner minutes per accepted output, including corrections.
- End-to-end throughput, memory pressure, failure rate and recoverability.
- Narration intelligibility, diagram legibility and usefulness against predefined criteria.
- Licensing/provenance completeness.

Demand, audience retention and conversion remain **unmeasured** without appropriate audience evidence. Private quality checks cannot validate them.

### First no-publication milestone

Produce a **private feasibility dossier**, not a channel launch. After explicit authorization for local testing, use a small, predefined set of topics in one niche to assess the proposed workflow. Before testing, freeze acceptance criteria and stopping conditions.

The dossier should contain:
- Finished private examples with reproducible demonstrations.
- Asset/model license and provenance records.
- Error and correction logs.
- Separate accounts for cash expenditure, owner time and hardware economic charge.
- A comparison with a simpler manual or partially automated workflow.
- A go/revise/stop decision.

Stop or redesign if correctness, differentiation or owner-time requirements fail. Passing this milestone would establish limited production feasibility—not demand or revenue. No execution is authorized by this review.

### Keep the “tower” explicitly provisional

- **Business pseudo-instance:** authored hypotheses about audience, value and revenue—not a discovered opportunity.
- **Process:** proposed editorial, review and support responsibilities with unresolved legal and eligibility gates.
- **Flows:** proposed production sequences, budgets, failure handling and human checkpoints.
- **Composites:** candidate capabilities such as “verified demonstration” or “narrated explainer,” accepted only against explicit contracts.
- **Components/configuration:** actual tools, models, licenses and settings, grounded individually before composition.

Compose upward only when evidence supports the next contract. Preserve failed assumptions and alternatives. **A connected pipeline is not an emergent business, and a renderable video is not a commercially useful product.**
