## Make retrieved capabilities useful without pretending they already form a system

**Abstract.** Paper 10 retrieves source-backed capabilities; a list of hits is not an architecture. This paper implements a proposal layer that retains requirement alternatives, compares retrieved options, rejects known hard-constraint conflicts, and records provisional choices with their actual archived contracts. An authored media profile proposes semantic dependencies separately from original source wiring. ## 1. Question, scope and falsifiable claims

**Question:** Can source-backed retrieval support an inspectable architecture hypothesis while preserving requirements, conflicts and uncertainty?

| Hypothesis | Check | Failure condition |
|---|---|---|
| H1: requirements survive selection | Every requested capability retained; alternatives represented as OR groups | A missing capability disappears or both alternatives become mandatory |
| H2: candidate decisions are justified | Known conflicts and structurally unresolved sources rejected; reasons retained | A cloud-bound option silently satisfies local-only |
| H3: source interfaces are grounded | Native field/port inventories and source hashes verified | An invented port, source or wire is presented as recorded |
| H4: semantic mapping stays explicit | Authored relations separated from archived wire witnesses and type overlaps | Message-to-Message is treated as proof of intended behavior |
| H5: revision is non-destructive | Parent digest and full prior proposal preserved | An exclusion overwrites the original or silently invents a replacement |

**Alternatives:** unqualified first-hit selection may be faster but can violate hard constraints; a human may produce better architecture hypotheses; the authored relation profile may be wrong or incomplete. There is no independent architecture-quality or runtime-success claim.


## 2. Proposal data model and evidence levels

| Record | Content | Evidence boundary |
|---|---|---|
| Capability slot | Original request, state, primary hypothesis and alternatives | A slot is not a new executable composite |
| Requirement group | Required capability or OR alternative | Provisional coverage is not validated completion |
| Candidate | Source, ports, settings, dependencies, assessment and rejection reasons | Archived declaration and configuration, not execution proof |
| Semantic interface hypothesis | Domain input/output meaning | Authored profile, not an exported native port |
| Relation | Hypothesized source/target roles, possible port pairs and archived witnesses | No automatic wiring or converter is implemented |
| Revision | Parent reference, reason, digest and preserved history | Checksums are integrity checks, not authentication signatures |

The authored media profile is limited to the named media capabilities. Other domains keep their retrieved slots but receive a missing-relationship-profile gap. No order or wiring is invented for them. Unknown local/Mac compatibility remains provisional. Empty dependency metadata means none declared, not no dependencies.


## 3. Selection and connection procedure

1. Ask Paper 10 for capability-specific candidates and constraint evidence.
2. Retain all requested slots and OR alternatives.
3. Reject explicit exclusions, known runtime conflicts, structural issues and unknown relevance. A partially evidenced connected subset cannot silently stand for the whole capability.
4. Propose an admissible source-backed option, explicitly provisional; keep rejected and alternative options with reasons.
5. Extract actual archived port declarations and unconnected settings; public composite interfaces remain unconfirmed.
6. Apply the authored media relation profile only where applicable. Check literal port-type overlaps and separately inspect bounded shared source graphs for original edges.
7. Preserve missing endpoints and unmapped interfaces. Do not invent an adapter, semantic compatibility, hardware fit, accepted architecture or executable result.

This procedure does not maximize apparent coverage. Refusing a conflicting option is preferable to filling a slot incorrectly.

## 6. Revision and restoration

This controlled exclusion tests revision handling; it is not an observed defect in the candidate. The same hard constraints remain in force. The child either proposes a different admissible option or leaves the slot unfilled. Saved JSON outputs retain history; restoration validates the original contracts and parent chain against current source evidence. No separate remote service or database is introduced.


```python
before=study_designer.revision(1)
excluded=next(s['primary']['id'] for s in proposal['slots'] if s['capability']=='script_input')
child=study_designer.propose(VIDEO_QUERY,k=6,parent=1,excluded_ids=[excluded],revision_reason='Controlled candidate exclusion to test revision; no observed defect claimed')
assert study_designer.revision(1)==before
assert child['parent_digest']==before['digest']
restored=ArchitectureDesigner.restore(study_designer.history(),study_designer.engine)
assert restored.revision(1)==before
print('Parent preserved and two-revision history restored.')
display(JSON({'controlled_child':child,'complete_saved_history':study_designer.history()},expanded=False))
```

## 8. Counterarguments and unresolved design obligations

A capability name or purpose statement can overstate what its source really does. A native field named “Data” can carry incompatible content; literal type overlap is only a candidate mapping. Archived source wiring may have been configured for another purpose. Source permissions, current component versions, cost, model/hardware fit and behavior still need qualification. A local-video requirement cannot be satisfied merely by finding cloud video tools.

The profile's proposed audio-to-subtitles dependency does not assert that the retrieved subtitle tool accepts raw audio: transcription or another explicit adapter may be required. Its actual recorded inputs must be mapped before any build. The system never selects an invented converter to hide that gap.

## 9. Conclusion and next phase

Paper 11 turns retrieved evidence into a structured, revisable architecture hypothesis. It does not turn partial evidence into a working instance. The observed local-video result has provisional pieces and real unfilled requirements; that is useful design information. The next phase should qualify a specific missing capability or semantic interface against real source code and runtime evidence before attempting assembly.
