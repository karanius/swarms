# Iterative Instance Revision: From a New Requirement to an Updated Design — Paper 5

## Abstract

This implementation study investigates a bounded revision algorithm for an existing canonical instance. A structured policy proposal is mapped to an explicit requirement realization, propagated through declared upward membership and downstream data/lifecycle relations, and converted into a narrow patch. Existing identities, native component references, unrelated configurations, requirements, questions and unresolved gaps must remain preserved. A notebook comparison examines impact closure, a direct-only ablation and an unselective rewrite on authored cases. The algorithm is integrated into V3 as preview, inspect and save-as-proposed controls with base-revision and digest guards. The example business rule is not automatically approved; raw-ticket interpretation and behavioral evidence remain explicit gaps. Results below are written from executed local experiments. This is not a free-text specification interpreter, autonomous new-component generator or self-improving algorithm.


## 1. Objective and Relationship to the Machine

**Research question:** Can the machine incorporate one new structured requirement into an existing instance, discover its declared graph impact, and produce a validated revision while leaving unrelated design intact?

Paper 1 defined descriptions, identities and preservation. Paper 2 described boundary resolution over versioned obligations. Paper 3 supplied a narrow structural assembly demonstration. Paper 4 implemented more of the bounded controller, including declared refinement and alternative restoration. Paper 5 now exercises **instance evolution**: the instance is the running example, while the algorithm and its revision procedure are the research deliverable.

| Objective | Delivered mechanism | Boundary |
| --- | --- | --- |
| Accept a new design condition | Explicit structured policy grammar and target requirement ID. | No unrestricted natural-language understanding. |
| Find what it touches | Direct realization seed plus graph closure. | Declared structural reachability, not complete semantic causality. |
| Revise incrementally | Append a proposed requirement, targeted instruction and reassessment references. | Does not rebuild or fill every layer. |
| Preserve unaffected work | Field-level comparisons and native/reference validation. | Tests cover declared invariants, not every possible defect. |
| Keep iteration inspectable | Preview, patch inspection, immutable revisions and retained open gaps. | No claim that a proposed policy is correct or approved. |

The algorithm grows the instance's description. It does not learn new code or change its own grammar. The two supported ticket facts are a small domain adapter; the impact traversal and preservation procedure are reusable over the declared graph relations. Whether a broader adapter is useful remains future work.

## 3. Revision Algorithm and Invariants

### 3.1 Impact rule

Let V be occurrence/design identities. Define a review relation E from (a) a lower member to its containing higher representation, and (b) the source of a declared data or lifecycle link to its target. Resource dependencies are preserved but are not traversed for this policy-change adapter. A future resource-change adapter would need different propagation rules.

From the unique direct realization seed S, compute A₀ = S and Aₜ₊₁ = Aₜ ∪ successors_E(Aₜ) until no identity is added. The implementation uses a queue and visited set, recording the first discovery reason for each identity. On a finite resolved graph, each identity enters the queue at most once. This bounds visits by |V|; it does not prove that E captures every semantic effect.

The current implementation scans memberships/links while processing each identity. Its work is bounded by the examined graph and is approximately O(|A| × (|V| + |M| + |L|)), where M is membership incidence and L is the link list. It is not advertised as an optimized adjacency-index implementation. Impact and patch-size limits gate acceptance after construction; they are not a hard wall-clock sandbox.

| Transition | Written result | Guard |
| --- | --- | --- |
| Normalize | Supported rule record. | Known grammar and stable identity. |
| Resolve target | One explicit prompt-bearing realization. | Reject missing or ambiguous target. |
| Check prior policy | Duplicate/conflicting condition disposition. | No silent replacement of existing rules. |
| Compute impact | Seeds, affected/unchanged sets and reasons. | Referenced neighbors resolve; each visited once. |
| Propose patch | Append requirement; append direct instruction; mark reachable items for reassessment. | Preserve existing member identities and connections. |
| Retain uncertainty | Approval, predicate interpretation and behavior gaps. | No generated evidence or policy confirmation. |
| Validate | Preservation plus Ajv envelope/native-reference checks. | Failed candidate cannot be committed through the live endpoint. |
| Save | New immutable revision and atomic head update. | Exact base revision and regenerated plan digest. |

### 3.2 What changes and what does not

| Region | Allowed change for this adapter | Preserved |
| --- | --- | --- |
| Direct assessment occurrence | Append proposed policy instruction and policy reference. | Native definition/version, ports, other configuration fields and identity. |
| Reachable downstream/upper items | Add policy reference and reassessment marker. | Their configurations and membership/interfaces. |
| Unreachable items | None. | Entire entity records. |
| Requirements | Append one proposed child requirement. | Every previous requirement and its discovery status. |
| Gaps | Append policy approval, predicate-contract and behavioral obligations. | Existing unresolved gaps and dispositions. |
| History | Append algorithm-proposal decision. | Prior decisions. |
| Questions, links and library | None. | No fabricated user answer, rewiring or native-definition edit. |

Reachability deliberately overapproximates the review scope. Marking a draft output or process for reassessment is not the same as rewriting its behavior. Only the direct prompt changes; instruction text alone is not behavioral proof.


## 4. Study Protocol, Comparisons and Revision Log

Fourteen cases are authored from the captured instance: five supported variations and nine blocked conditions. The same assistant authored algorithm, cases and assessor. The initial integration preview preceded the protocol. This is an exposed development study, not preregistered or independently held-out evaluation.

| Cases | Purpose |
| --- | --- |
| Base outage and data-loss rule | Apply the same mechanism to either supported fact. |
| Unrelated extra item | Ensure an additional disconnected occurrence remains unchanged. |
| Disconnected downstream | Ensure impact shrinks when the corresponding native/projected data links are removed. |
| Additional consumer | Ensure an explicitly connected extra consumer enters the review scope. |
| Missing/ambiguous realization | Return the policy-adapter gap instead of choosing arbitrarily. |
| Conflict or duplicate rule | Preserve prior policy and require a separate decision/versioning operation. |
| Unsupported field/operator | Return a grammar gap. |
| Impact/patch limits | Return a bounded blocked proposal, not a partial policy disguised as complete. |

| Method | Difference | Live-commit status |
| --- | --- | --- |
| impact | Full declared closure, selective patch. | Only this method is exposed for commit. |
| direct-only | Updates seed without upward/downstream closure. | Experiment ablation only; preservation gate should reject omitted impact. |
| rewrite-all | Marks every item, including unrelated ones. | Experiment ablation only; preservation gate should reject unnecessary changes. |

An independent code-level assessor uses separately authored expected affected sets and field comparisons. It does not establish independent authorship. The implementation's preservation guard is also exercised with deliberately corrupted candidates. All failed comparisons remain in the evidence.

| Criterion | Decision |
| --- | --- |
| Reference match | Exact expected blocked reason, or affected-set match plus candidate/reference checks. |
| Preservation | Prior requirements/gaps/history, identities, links, questions and unrelated/nonseed configurations remain intact. |
| Approval/evidence integrity | New rule remains proposed; predicate and behavior gaps stay open. |
| Save safety | Preview does not persist; stale or tampered plans are rejected. |

After the initial comparison, the guard was strengthened to retain previous decision records and unchanged entity count. Round-1 observations are preserved. The same exposed cases are rerun below, with source/protocol snapshots saved before each run. No independent replication is claimed. During UI inspection, the user identified confusing role labels on the Components plane; the display now leads with Chat Input, Agent, Agent and Chat Output, keeping instance roles separate from native component identity. That is a representation correction, not a change of layer or component implementation.

```python
from pathlib import Path
import subprocess, json, shutil
from IPython.display import display, Markdown
repo = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "knowledge/research/paper-5-study/run-study.mjs").exists()), None)
if repo is None:
    raise FileNotFoundError("Keep the study modules with this repository notebook.")
node = shutil.which("node")
if not node:
    raise RuntimeError("The local Node.js runtime is required to run the shared revision algorithm.")
completed = subprocess.run([node, "knowledge/research/paper-5-study/run-study.mjs"], cwd=repo,
                           text=True, capture_output=True, timeout=60, check=True)
study = json.loads((repo / "knowledge/research/paper-5-study/observations/results.json").read_text())
cols = ["method","cases","passes","proposed","preservationRejected"]
display(Markdown("| " + " | ".join(cols) + " |\n|" + "|".join(["---"]*len(cols)) + "|\n" +
                 "\n".join("| " + " | ".join(str(r[k]) for k in cols) + " |" for r in study["summary"])))
print("Study acceptance:", json.dumps(study["checks"], sort_keys=True))
assert all(study["checks"].values())
```

```python
import matplotlib.pyplot as plt
rows = study["summary"]
fig, ax = plt.subplots(figsize=(8,4), constrained_layout=True)
labels=[r["method"] for r in rows]
passes=[r["passes"] for r in rows]
misses=[r["cases"]-r["passes"] for r in rows]
ax.bar(labels,passes,color="#2f8d81",label="Reference matched")
ax.bar(labels,misses,bottom=passes,color="#b75a5a",label="Reference missed")
ax.set_ylabel("Authored cases"); ax.set_yticks(range(0,15,2)); ax.legend(loc="lower left")
ax.set_title("Structured revision comparison · same-author exposed suite")
plt.show()

example=next(r for r in study["records"] if r["case"]=="base-outage" and r["method"]=="impact")
impact=example["plan"]["impact"]
baseline=json.loads((repo/"knowledge/research/paper-5-study/baseline.instance.json").read_text())
lookup={e["id"]:e for e in baseline["entities"]}
display(Markdown("| Affected item | Reason | From |\n|---|---|---|\n" + "\n".join(
    f"| {lookup[i]['name']} | {impact['reasons'][i]['kind']} | {impact['reasons'][i]['from'] or 'target requirement'} |"
    for i in impact["affectedIds"])))
display(Markdown("**Unchanged:** " + ", ".join(lookup[i]["name"] for i in impact["unchangedIds"])))
```

```python
from matplotlib.patches import FancyBboxPatch
layers=sorted(baseline["layers"],key=lambda l:-l["order"])
positions={}
for row,layer in enumerate(layers):
    for col,e in enumerate([e for e in baseline["entities"] if e["layer"]==layer["id"]]):
        positions[e["id"]]=(col*2.5, -row*1.65)
fig, ax = plt.subplots(figsize=(12,8), constrained_layout=True)
for e in baseline["entities"]:
    for child in e["members"]:
        x,y=positions[e["id"]];xx,yy=positions[child]
        ax.plot([x+.95,xx+.95],[y-.2,yy+.6],linestyle=":",color="#8291a1",zorder=1)
for link in baseline["links"]:
    if link["kind"]!="data": continue
    x,y=positions[link["from"]];xx,yy=positions[link["to"]]
    ax.annotate("",xy=(xx,yy+.2),xytext=(x+1.9,y+.2),arrowprops=dict(arrowstyle="->",color="#66798c"),zorder=1)
for row,layer in enumerate(layers): ax.text(-1.9,-row*1.65+.2,layer["name"],va="center")
for oid,(x,y) in positions.items():
    color="#bce6da" if oid in impact["seeds"] else ("#f0dfba" if oid in impact["affectedIds"] else "#dce5ef")
    ax.add_patch(FancyBboxPatch((x,y-.2),1.9,.8,boxstyle="round,pad=.04",facecolor=color,edgecolor="#8391a2",zorder=2))
    e=lookup[oid]; label=(e["itemRef"]["id"] + "\n" + e["name"]) if e.get("itemRef") else e["name"].replace(" ","\n",1)
    ax.text(x+.95,y+.2,label,ha="center",va="center",fontsize=8,zorder=3)
ax.set_xlim(-2,10);ax.set_ylim(-7.6,1);ax.axis("off")
ax.set_title("Recorded impact: direct configuration (green), reassess (amber), unchanged (blue)")
plt.show()
display(Markdown("| Corruption | Detected preservation failures |\n|---|---|\n"+"\n".join(
    f"| {r['name']} | {', '.join(r['failures'])} |" for r in study["negatives"])))
print("Cycle helper scope:",study["helper"])
```
