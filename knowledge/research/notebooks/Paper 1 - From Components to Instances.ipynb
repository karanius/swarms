# From Components to Instances: Iterative Composition

## Abstract

This conceptual paper defines a descriptive framework for generating an intent-specific instance specification from versioned library items. It separates an item description from the rule that operates on descriptions, and separates horizontal assembly from vertical representation. Rules propose typed arrangements, preserve member identities and interface mappings, and carry unresolved requirements forward explicitly. The intended product is a multilayer JSON specification, not a running system. We define the representation, rule judgments, and obligations that a future generator must respect; we do not claim a general synthesis algorithm, behavioral correctness, or empirical superiority.

## 1. Introduction and Scope

An instance is a purposeful vertical slice of a layered library: it selects, configures, and connects only the pieces needed for a stated intent. At day zero, executable components may already exist while higher-layer items do not. The research question is how structured descriptions and explicit rules can write the missing assemblies without inventing unsupported capabilities. This first paper establishes the vocabulary and rule semantics. Candidate search and layer-specific policies are reserved for Paper 2. Runtime implementation and evaluation are later work.

## 2. Context and Related Foundations

Langflow documents inputs and outputs as configurations governing component data flow, editor fields, ports, and connection validation [1]. This provides concrete interface facts, but does not establish a general algorithm for satisfying arbitrary intents across our library layers. JSON Schema specifies structural assertions over JSON documents [2,3]; passing those assertions does not by itself establish the meaning or correctness of a generated system. We use these sources as implementation context, not as evidence that the proposed framework works. This is a selective conceptual grounding, not a systematic literature review or a claim of novelty.

## 3. Component Descriptions: The Genetic Makeup Metaphor

For this paper, genetic makeup is a metaphor for a structured description, not a biological or learning mechanism. Write d = (k, ℓ, C, I, O, Θ, D, E). Here k is a stable definition identity plus version; ℓ is its library layer; C describes capability and limits; I and O describe ports and conditions; Θ is the configurable parameter space; D lists dependencies; E records evidence and its scope. Identity and layer are metadata around the six descriptive fields below. Natural-language descriptions may be retained, but a future generator must distinguish explicit machine-checkable constraints from uninterpreted claims. Definition keys are namespace-qualified: a description can belong to a pinned library snapshot or remain a proposed, instance-local definition. Evidence records identify the definition, configuration and dependency versions they concern; they do not automatically remain applicable after a change.

| Field | Required description |
| --- | --- |
| C — Capability | Purpose, intended behavior, limits, and any formal claim or natural-language qualification. |
| I — Inputs | Named ports, accepted types, preconditions, required or optional status. |
| O — Outputs | Named ports, produced types, postconditions, and the status of guarantees. |
| Θ — Configuration | Declared parameters, permitted values, defaults, and cross-parameter constraints. |
| D — Dependencies | Versioned item and resource references, transitive prerequisites, and unresolved bindings. |
| E — Evidence | Claim, tested or analyzed version, method, result, provenance, and applicable scope. |

## 4. Library Definitions, Instance References, and Layer Graphs

Let Lᵥ be a versioned library snapshot, and Λ the ordered set of library layers. An instance node is an occurrence with its own identity, a reference to a definition version, and a configuration binding. Two occurrences may reference the same definition without being the same instance. Each relevant layer has a typed attributed graph Gℓ = (Vℓ, Eℓ). Higher-layer nodes reference the same lower-level occurrence identities they represent; they do not silently duplicate them. A membership relation records this vertical representation. A layer may be empty or marked unnecessary with a reason. No rule requires filling every layer merely because the layer exists.

## 5. Intent, Requirements, and the Generated Instance Document

Intent is the initial statement of purpose; requirements refine it through discovery. Define the document J = (intent, R, {Gℓ}, P, B, U, T), where R is the current requirement set, P maps higher items to lower members and interfaces, B contains configuration and resource references, U contains unresolved gaps, and T records versions, decisions, and supporting judgments. Requirements have identity, scope, mandatory or optional status, and acceptance criteria. Configuration references below Components remain part of the description: model access, storage, tools, and credentials may be required for later activation, but no live authentication is needed merely to describe the intended binding. Secrets are referenced rather than embedded. Acceptance is judged against a versioned requirement baseline Rᵦ with explicit scope membership. Every mandatory requirement in that scope must retain a judgment and its evidence criterion. Removing a requirement or narrowing scope is an explicit versioned decision, not an unexplained increase in progress. U contains active unresolved obligations; rejected candidates and decision history belong in T. Updates to U identify which obligations are added, resolved, superseded, or retained.

## 6. Generation Rules as Operators over Descriptions

Distinguish a description dₖ from an occurrence v = (o, k, θ), where o is an occurrence identity, k references a scoped definition version, and θ binds its parameters. A rule declares its input pattern, applicability conditions, construction, outputs, and remaining obligations. Write r(Rᵦ, G, S, θ, W) as proposing (G꜀, d*, v*, M, U′, T′). G is the surrounding graph; S identifies selected occurrences, including any explicitly declared new occurrences; W records intended connection edits. The construction first writes candidate graph G꜀ with those edits and records the treatment of existing incident edges. It then identifies the member subgraph H = (S, F), including its retained internal edges, and computes boundary incidence from G꜀ rather than guessing it from F. The result d* is a proposed instance-local assembly description, and v* is its higher-layer occurrence. Registering a reusable library definition is a separate, explicit operation. This is a rule interface, not a solver implementation or a guarantee that a satisfactory proposal exists.

| Rule field | Declared meaning |
| --- | --- |
| Input pattern | Requirements and a selected set of versioned item occurrences. |
| Conditions | Typed endpoints, semantic assumptions, valid configuration, dependency closure, and layer-specific constraints. |
| Construction | A declared horizontal graph, configuration bindings, and external boundary mapping. |
| Output | A proposed higher-layer description referencing the same member occurrences. |
| Residual obligations | Unknown conditions and evidence requirements, each with identity, status, and requirement references. |
| Trace | Rule identity/version, sources, rationale, rejected alternatives, scope changes, and affected requirements. |

## 7. Applicability Judgments: Met, Violated, or Unknown

A rule evaluates each relevant condition as met, violated, or unknown under stated evidence criteria. Met means the available description and evidence establish that particular condition. Violated means available information contradicts it. Unknown means the necessary information or justification is absent or insufficient. Unknown must not be coerced into met. A violated mandatory condition rejects that candidate unless an explicit alternative or adapter changes the proposal. An unknown design condition keeps the proposal conditional and records a gap. These judgments apply to types, semantics, configuration, dependencies, requirements, and evidence separately. Judgments themselves record the requirement baseline, rule version, description versions, evidence references, and applicable assumptions. If evidence both supports and contradicts a condition, preserve that conflict in the record; the condition cannot count as met until the conflict is resolved under declared criteria. The outward judgment remains unknown while the conflict is unresolved. Changed configuration or dependencies trigger an applicability review of earlier judgments.

## 8. Horizontal Assembly and Deriving a Boundary

Horizontal assembly proposes and records connections among occurrences within a layer. A connection judgment checks named endpoints, accepted and produced types, required fields, configuration constraints, dependencies, and declared behavioral conditions. Under a narrow nominal type rule, an output type may connect when explicitly accepted by the target input; conversions require a named adapter or separately described binding. The selected assembly H includes both its member occurrences and retained internal edges F. Its boundary records required unbound inputs, intentionally exposed ports, and every edge crossing the selection in candidate graph G꜀. Exposed parameters and dependencies are recorded as well. These judgments license a structural proposal, not a behavioral guarantee. State, ordering, failure propagation, cycles, and resource competition can change an assembly’s meaning.

## 9. Vertical Packaging and Identity Preservation

Vertical packaging writes a higher-layer occurrence representing H and referencing its instance-local description. In Paper 1 it is purely representational: it does not create a runtime or change execution semantics. Preserve member identities, pinned definitions, bindings, and retained internal edges. For each crossing edge in G꜀, retain its original member endpoint and exposed-interface mapping. Several external edges may use one exposed port, but distinct internal endpoints must not be silently conflated. Required unbound inputs remain visible obligations rather than disappearing because no edge crosses them. Unfolding follows the retained references to recover H and the incident boundary relations of G꜀; it does not undo intentional edits that produced G꜀ from G. This preservation claim assumes complete and unchanged records. Overlapping wrappers may reference the same stateful occurrence: shared identity does not imply ownership, concurrency safety, resource isolation, or absence of interference.

## 10. Recursive Description without Assuming Universal Layer Semantics

The proposed higher item receives the same general description fields: claimed capability, exposed inputs and outputs, configuration surface, dependencies, and evidence or obligations. This permits another rule to consider it as an operand. What changes between Composite, Flows, Processes, and other layers is not merely a numeric level. Each layer needs its own admissible objects, composition meaning, and acceptance conditions. Those definitions cannot be inferred from a label or a node count. This paper supplies a common record structure, while Paper 2 must specify how candidate rules are selected and how layer-specific constraints are represented. Recursive use refers to the proposed scoped description and occurrence; copying that description into the shared library requires separate registration and its own evidence/status record.

## 11. Worked Symbolic Example: A Proposed Summarizer

Consider three versioned Langflow descriptions: Chat Input provides a Message, Agent accepts a Message and declares a Message response, and Chat Output accepts a Message. The connected local catalog supports both proposed edges: ChatInput.message → Agent.input_value and Agent.response → ChatOutput.input_value. A rule may therefore write this candidate graph under the stated nominal type rule. To describe a three-point summarizer, it must additionally assign instructions, a model reference, output-format obligations, grounding requirements, and failure conditions. Port compatibility alone establishes none of those behavioral claims. Packaging writes a proposed Summarizer item whose members are those same occurrences; its input is source text, its output is a Message, and its task-level evidence remains separately stated.

| Connection | Declared interface judgment | Not established by this judgment |
| --- | --- | --- |
| Chat Input → Agent | Message is explicitly accepted. | Accuracy, source grounding, and useful summarization. |
| Agent → Chat Output | Message is explicitly accepted. | Successful runtime, correct failure handling, or requirement satisfaction. |

## 12. Worked Boundary and Unknowns

The proposed external source-text entry is a binding to Chat Input’s input_value string field. This is distinct from the Message-valued edge leaving Chat Input. The record must identify that field binding and any required construction or conversion; Message compatibility does not establish it. The external exit maps to Chat Output’s message port. Named model bindings and optional resource references remain explicit. A missing provider reference can be written as an unresolved dependency and must not be presented as operational readiness. A source-grounding requirement without sufficient support remains an obligation, not a guarantee. Reuse above this boundary is a scoped proposal and does not certify every future use.

## 13. Iteration, Completion, and Stopping

Design can proceed downward from requirements, upward from existing items, or from an intermediate boundary. Each attempted transition identifies a gap, proposes a rule application, records its judgments, and updates affected graphs and mappings. Progress is relative to the unchanged requirement baseline unless a scope change is explicitly recorded. Structural completeness, satisfaction, and execution readiness are independent judgments. A structurally complete record may contain unknown behavioral obligations. Satisfaction for a declared scope requires a supported met judgment for every mandatory requirement under its evidence criteria; unknown or violated conditions preclude that satisfaction claim. Rejection concerns a candidate, not the impossibility of the intent. A blocked result identifies the unmet prerequisite or violated constraint for the current proposal or declared search horizon. A partial result retains unresolved obligations. Every attempted transition, including retries and failed expansions, consumes a positive unit of a finite budget, and suboperations must terminate or be interruptible. Exhaustion returns the current partial record and remaining obligations. This bounds the proposed procedure; it guarantees neither that a solution exists nor that search will find one.

## 14. Design Invariants and Counterexamples

The proposed generator must preserve reference identity and version pins, map every declared boundary connection, avoid unresolved endpoint references, retain the status of unknown and violated conditions, and trace higher-layer claims to requirements and obligations. These are specification obligations to be checked, not experimental results of this paper. Counterexamples explain their necessity: a type-compatible agent may answer the wrong question; individually valid components may interact incorrectly; duplicating a stateful member can change the system; a wrapper may lose a hidden dependency; a feedback loop may require ordering and termination semantics absent from a DAG. No unrestricted acyclicity assumption is imposed on all systems. A future rule must either model a cycle explicitly or declare it unsupported.

## 15. Historical Preliminary Evidence and Its Limits

Before the current design-only scope was established, a fixed recipe was recorded using real Langflow definitions. Five live short-source cases returned three nonempty bullets and received independent AI-agent source-grounding review; two invalid cases were blocked by evaluator guards before runtime. There were four distinct short sources and one repeat. These observations illustrate one authored recipe and an evidence-recording path. They do not validate rule discovery, general search, long-document summarization, semantic inference, or arbitrary layer generation. No additional runtime experiment is part of finishing this manuscript. The archived artifact and review remain available under the experiment record cited below. The evaluator files retain their original pending-review annotation; subsequent fixture-scoped judgments are recorded separately in review.json. They are not retroactively relabeled as generator self-assessments.

## 17. Discussion and Limitations

This framework makes proposed assemblies inspectable before they run. Its principal contribution is the separation of representations, rule judgments, and evidence obligations. Its limits are substantial: natural-language capabilities may be ambiguous; semantic implication can be difficult or unavailable; layer meanings are not yet fully defined; dependency closure can reveal missing resources; and several distinct assemblies may satisfy the same description. No optimal objective, universal composition rule, learned selection mechanism, or complete synthesis method has been established. Deterministic, neural, language-model, and hybrid approaches remain candidates to compare after their task and evaluation conditions are specified. Langflow can supply real component definitions and later serve as a translation target; this does not imply it already generates our entire layer hierarchy.

## 18. Conclusion and Handoff to Paper 2

Paper 1 defines what the generator must read, what a rule must declare, what a proposed higher item must preserve, and how unresolved conditions remain visible. It establishes a conceptual description-and-rule framework, not a proven universal generator. The finished manuscript provides a bounded foundation for the next paper: candidate selection and boundary-resolution policy, together with the first explicit layer-specific composition rules. Paper 2 should specify how to choose and revise a rule application, its search state and objective, how alternative proposals are compared, and when to return a complete, partial, or blocked specification. Execution and empirical comparisons remain deferred until the theoretical design is ready.
