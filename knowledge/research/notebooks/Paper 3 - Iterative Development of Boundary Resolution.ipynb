The descriptions may originate in a pinned Langflow catalog, but a catalog entry does not automatically supply behavioral contracts or our higher-layer semantics. Preparation must record any mapping from Langflow fields to these study fields, including omissions. Resource records may describe models, storage, tools or services below the Components plane. They contain authorized binding references and declared properties, never secret values. Writing a resource reference is distinct from authenticating to it, checking its live availability or calling it. This initial design task does not require those activation actions.

### 4.5 Rules, profile and bounded search policy

| Selection policy | Guard interpretation, boundary priorities, candidate ordering and tie-breaks. | Policy changes receive new versions. |

| Repeat handling | Semantic-state and attempted-action keys, including deliberate occurrence multiplicity. | Layout changes cannot masquerade as new search progress. |

No numerical limits are selected here. Later sections must choose them before evaluation and state which resources are matched across comparisons. Any condition permitting external retrieval must identify that access and preserve the returned material as part of the observation context.

### 4.8 Composite representation and interface boundary

A composite output must contain more than a name and a list of nodes. It represents a selected lower subgraph inside the proposed candidate context.

| Composite record | Required content |

An external text field binding and a Message-valued data edge are distinct relations; the schema must not collapse them into one untyped connection. ### 4.9 Outcome and observation status
