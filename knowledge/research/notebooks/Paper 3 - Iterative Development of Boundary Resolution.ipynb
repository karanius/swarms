## Abstract

This study investigates a bounded structural implementation of boundary resolution for Components → Composite specification generation. Three deterministic methods are compared on three development and fourteen same-author designated evaluation cases. Explicit nominal interfaces, constraints and packaging rules produce either a proposed composite or an unresolved disposition. The repaired procedure passes all seventeen authored reference assessments, producing eleven accepted artifacts and six non-construction outcomes; comparison methods expose invalid candidate choices and omitted required inputs. These are descriptive results from a small synthetic suite with a planned correction, a weak baseline and no independently held-out evaluation. They demonstrate an executable starting point, not full implementation of the general controller, behavioral correctness or generation across all layers. Code, JSON observations, tables and figures are retained in the notebook and accompanying study artifacts.


## 1. Research Question and Scope

**Research question:** Given confirmed requirements and a library of component descriptions, can boundary resolution produce a coherent composite specification—or identify exactly what prevents it?

The initial study is limited to the **Components → Composite** transition. Its output is an assembly specification, rather than a deployed system. Confirmed requirements and component descriptions are supplied as inputs; eliciting requirements through conversation is outside this initial scope.

| Assessment criterion | What we will examine |
| --- | --- |
| Requirement coverage | Whether every supplied requirement is traced to a proposed realization or remains explicitly unresolved. |
| Connection validity | Whether proposed connections respect declared interfaces and configuration constraints. |
| Gap reporting | Whether missing information, missing capabilities, and incompatible requirements are identified without being silently assumed away. |

These are proposed assessment criteria, not established results. Structural compatibility alone does not demonstrate that an assembly satisfies its behavioral requirements. Detailed measures and experimental cases will be developed in later sections.


## 2. Relationship to Papers 1 and 2

### 2.1 Purpose of the three-paper sequence

Paper 1 defines what an instance description and a valid composition record must contain. Paper 2 proposes how a bounded controller chooses its next revision when required properties and available pieces do not yet match. Paper 3 investigates an implementation of that proposal through recorded code experiments. These contributions are complementary: a representation does not supply a search strategy, and a search strategy does not establish that its generated system behaves correctly.

The immediate object of study is a **written Components → Composite specification**, not a deployed agent, business, or complete hierarchy. The broader ambition remains generation across the layers an intent needs. Evidence for the first transition will not, by itself, justify extending the same claims to Flows, Processes, or higher planes.

| Paper | Main question | What it contributes | What it does not establish |
| --- | --- | --- | --- |
| Paper 1 — From Components to Instances | What must a description, assembly, and higher representation preserve? | Description fields; definition/occurrence distinction; interfaces; obligations; identity and boundary preservation. | An implemented solver, behavioral correctness, or a universal generation rule. |
| Paper 2 — Boundary Resolution | Given the current design and unresolved obligations, which move is permitted next? | Decision state, boundary selection, candidate guards and ordering, branch handling, finite limits, scoped outcomes. | Empirical effectiveness, optimal search, complete search, or semantics for every layer. |
| Paper 3 — this study | Can an implementation produce useful, inspectable specifications, and which rules need revision? | A planned sequence of implementation, development observations, explicit revisions, and separate evaluation. | Results before experiments, automatic generalization, or operational readiness from structural checks. |

**Conceptual dependency:** Paper 1 representation and preservation obligations → Paper 2 decision procedure → Paper 3 implementation and observations → proposed, versioned revisions. The final arrow returns proposed changes to the theory; it does not retroactively change the rules under which an earlier result was produced.

### 2.2 Translating the inherited descriptions into experimental inputs

Paper 1 describes an item as d = (k, ℓ, C, I, O, Θ, D, E). Identity and layer accompany six descriptive fields. “Genetic makeup” denotes this explicit description; it does not imply that useful compositions emerge automatically, that a biological process is simulated, or that a learned model has already been selected.

| Inherited field | Meaning in the initial implementation | Interpretation boundary |
| --- | --- | --- |
| k and ℓ | Pinned definition identity/version and layer. | A display label is insufficient to establish identity or legal placement. |
| C — capability | Stated purpose, limits, and qualified claims. | Natural-language similarity is a proposal signal, not proof of suitability. |
| I and O — interfaces | Named ports, accepted/produced types, preconditions and postconditions where supplied. | A compatible type establishes that interface condition only. |
| Θ — configuration | Parameters, bindings, defaults and declared constraints. | Writing instructions does not prove their behavioral effect. |
| D — dependencies | Item/resource references and unresolved prerequisites. | Naming a resource does not establish access, availability or authorization. |
| E — evidence | Support and contradiction records with version/configuration scope. | Missing evidence remains unknown; evidence does not transfer silently after change. |

An occurrence v = (o, k, θ) has its own identity o, references definition k, and supplies configuration θ. Two occurrences may use the same definition without becoming the same occurrence. Packaging must preserve the selected occurrence identities. Registering a reusable library definition remains a separate operation from creating an instance-local composite.

The component descriptions constrain candidate construction; they do not uniquely determine it. A rule registry and proposal mechanism must still supply operand choices, configuration choices, wiring and permitted refinements. The initial experiment design must identify that mechanism explicitly, including any authored recipe, so that recipe replay is not reported as autonomous assembly discovery.

### 2.5 Decision procedure inherited as a baseline proposal

| Decision stage | Inherited convention | What remains to investigate |
| --- | --- | --- |
| Extract boundary | Link an unresolved condition to its upper obligation and available or absent realization. | Whether the supplied descriptions expose the important mismatch. |
| Select boundary | Resolve prerequisites first; then use mandatory status, declared impact rank and stable order. | Whether that priority is useful on the chosen cases. |
| Generate proposals | Produce a finite, attributed batch from a declared registry/domain and caps. | Candidate coverage and the consequences of omissions. |
| Gate proposals | Reject violated mandatory guards; preserve unknowns according to structural/semantic class. | Correct implementation and adequacy of the guards. |
| Order candidates | Use the declared lexicographic policy, including conditional status and action rank. | Sensitivity to ordering; no optimality is assumed. |
| Apply and reassess | Accept one atomic design patch, retain residuals, invalidate affected judgments. | Whether recorded changes preserve the inherited obligations. |
| Return or branch | Use finite frontier/budget rules and explicit outcome records. | Usefulness of returned partial records and limits of the search horizon. |

Paper 2's action preference begins with specified reference resolution, reuse, composition and requested packaging, before supported decomposition and other declared revisions. Clarification is an information gate; it is not an arbitrary cheap candidate. This is an authored baseline convention, not an empirically discovered best policy. Deterministic, model-driven, neural and multi-agent proposal mechanisms remain possible alternatives; none is selected or validated by this section.

The finite-stopping argument is conditional: a finite charged-attempt budget bounds attempts only when each helper also returns or can be interrupted in finite time, and bookkeeping remains finite. It establishes neither solution discovery nor a practical latency bound. Iteration through controller revisions must also be distinguished from cycles inside an assembly graph. The initial conservative DAG profile does not provide semantics for arbitrary feedback systems.

## 3. Hypotheses, Alternatives, and Refutation

### 3.1 What this section commits us to

The study asks whether explicit boundary-resolution rules can help write a coherent Components → Composite specification from confirmed requirements and pinned component descriptions, or return a justified account of what prevents closure. This section separates three kinds of statement: implementation obligations, empirical hypotheses, and broader ambitions. None is reported as an experimental finding. No algorithm experiment is performed by writing this section.

| Statement class | Meaning | Appropriate treatment |
| --- | --- | --- |
| Conformance obligation | Behavior the implementation must satisfy to count as implementing the declared rule. | An observed violation is a defect or a specification problem; it cannot be averaged away by successes. |
| Empirical hypothesis | A prediction about usefulness or performance on a defined set of cases. | Investigate with recorded observations and an appropriate comparison. |
| Exploratory question | A consequential uncertainty for which the present theory gives no justified directional prediction. | Examine openly; do not retrospectively portray the outcome as predicted. |
| Long-term ambition | Generation across additional required layers and broader intents. | Retain as motivation; do not claim it follows from the initial transition. |

The unit of observation will be a case under an identified requirement baseline, library, rule/policy version, implementation, proposal source and resource limit. Several attempts on the same case are repeated observations of that case, not automatically independent evidence of generalization. The detailed case construction and evaluation protocol remain for later sections.

### 3.2 Central proposition and its boundaries

**Central proposition:** Within a declared finite composition domain, explicit descriptions and boundary-resolution rules can support the production of inspectable composite specifications, while preserving mandatory obligations and reporting unresolved conditions instead of inventing support.

This is a proposition to investigate, not a theorem established here. “Support” does not mean that the rules alone supply missing semantic knowledge. The candidate generator, registry, descriptions and judgment criteria remain necessary inputs. A hand-authored complete recipe may demonstrate representation or replay, but it cannot establish that the controller discovered an assembly.

| Term | Operational interpretation for this study | Excluded interpretation |
| --- | --- | --- |
| Coherent specification | References resolve; layer/profile constraints hold; selected connections and boundary mappings satisfy declared structural conditions. | The future system necessarily performs the intended task correctly. |
| Preserved requirement | Every mandatory scoped requirement retains a realization/judgment link or an explicit unresolved disposition. | A requirement is satisfied merely because its identifier remains in a file. |
| Justified gap | A specific unresolved condition with an affected requirement, grounds, context and next-information or capability need. | Every reported gap is real or every real gap has been discovered. |
| Useful output | An artifact or residual record that advances the stated design task under criteria fixed before its evaluation. | More nodes, fewer displayed gaps, or an attractive diagram alone. |
| Declared finite domain | Identified components, rule families, candidate/branch bounds and supported structural semantics. | Arbitrary programs, all Langflow components or all organizational layers. |

### 3.3 Conformance obligations that gate interpretation

The following are inherited requirements on the implementation. Passing them would establish conformance only for the examined artifacts and checks. Their interpretation must remain separate from any claim that the controller is useful or superior.

| ID | Obligation | Observable violation | Consequence |
| --- | --- | --- | --- |
| C1 | Mandatory requirement accounting survives accepted patches and branch changes. | A requirement disappears, loses its obligations, or is silently made optional. | The affected result cannot support a closure/satisfaction claim. |
| C2 | Unknown, violated and met judgments remain distinct and scoped. | Missing evidence, a proposal assertion or contradictory support is treated as met. | Invalidate the affected judgment and investigate its cause. |
| C3 | Packaging preserves member occurrence identities and complete external mappings. | Members are silently cloned; crossing endpoints or required unbound inputs disappear. | Reject the packaging result as nonconforming. |
| C4 | Accepted revisions satisfy the declared structural guards and layer profile. | An unresolved endpoint, illegal layer target or unsupported coercion is committed. | Reject that revision; do not compensate with an overall success rate. |
| C5 | Search limits and stopping dispositions are honored. | Budget is replenished by rollback, a helper cannot be interrupted, or exhaustion is reported as impossibility. | The finite-stopping premise or outcome interpretation is not met. |
| C6 | Results and reasons remain attached to their context. | A changed library/model/configuration inherits stale evidence, or rejected paths are erased. | The affected comparison is not adequately attributable or reproducible. |

An implementation defect and an inadequately specified rule can produce the same visible failure. Classification requires inspection of the frozen rule and trace. Until that diagnosis is supported, record the cause as unresolved rather than assigning it to whichever explanation protects the preferred theory.

### 3.4 Empirical hypotheses to investigate

The hypotheses below concern the initial scope. They do not yet contain numerical thresholds, sample sizes or statistical decision rules. Section 6 must define those before evaluation. Each hypothesis can be supported, weakened, contradicted within its scope, or remain inconclusive; absence of a detected failure is not universal confirmation.

| ID | Hypothesis | Predicted observation | Evidence that would challenge it |
| --- | --- | --- | --- |
| H1 — Grounded construction | On cases whose required structural assembly is expressible within the supplied descriptions, rules and declared horizon, the controller can construct an admissible composite specification without receiving the full answer as a recipe. | A generated artifact meets independently specified structural and requirement-accounting criteria. | Repeated failure on independently justified reachable cases, or success that depends on leaking the complete assembly into the input. |
| H2 — Informative gap reporting | When relevant information or capability is absent or conflicting, the controller can return specific, warranted residuals instead of unsupported closure. | Reported gaps identify affected obligations and documented grounds; assessable introduced gaps are detected without excessive unsupported reports. | Silent assumptions, false closure, missed consequential gaps, or vague/false gaps that prevent otherwise supported progress. |
| H3 — Added value of boundary control | Explicit boundary selection and obligation tracking improve the prespecified quality–resource trade-off relative to an appropriate comparison with the same available knowledge and access. | Better assessed artifacts/residuals under comparable limits, or equivalent quality with lower recorded resource use. | No meaningful advantage under the declared criterion, worse trade-offs, or an apparent gain explained by unequal information or resources. |
| H4 — Defensible iterative revision | Revisions motivated by development failures can improve the targeted weakness without unacceptable regressions on other criteria. | Improvement persists on fresh cases and earlier cases under an identified revised version. | Improvement is confined to the cases used to author the fix, or it introduces consequential regressions elsewhere. |

H1 needs independently justified reachability within the actual rule family and horizon. A compatible component list alone is insufficient. If reachability cannot be established, failure is informative but cannot fairly be classified as a missed known solution. Likewise, H2 cannot use the controller's own gap list as its reference standard. H3 remains a comparative hypothesis until the comparison and its fairness conditions are selected; H4 requires both development/evaluation separation and explicit regression criteria.

### 3.5 Alternatives and rival explanations

A successful example can have several explanations. The study must retain plausible rivals and collect enough context to distinguish them where possible. These are explanatory alternatives, not claims already demonstrated.

| Rival explanation | How it could mimic support | What a later study should distinguish |
| --- | --- | --- |
| Recipe replay | The answer is encoded in a template, example, component name or prompt. | Replaying a supplied solution versus selecting and wiring components from incomplete alternatives. |
| Catalog-specific shortcuts | A small catalog makes the intended nodes obvious. | Performance on structurally different cases and controlled irrelevant alternatives. |
| Proposer knowledge dominates | A model supplies the solution; the controller merely records it. | Proposal quality versus the additional effect of guards, boundary selection and tracking. |
| Unequal resources | One method gets more calls, time, context or hints. | Resource access and limits recorded for every comparison. |
| Permissive assessment | The checker tests formatting or types while ignoring the stated criterion. | Independently stated structural, coverage and semantic judgments. |
| Selective reporting | Only favorable cases or versions appear in the account. | Preserved case registry, failures, revisions and exclusion reasons. |
| Memorized development fixes | Revisions encode the observed examples. | Fresh evaluation cases separated from rule development. |
| Trivial conservatism | Returning unknown for everything avoids false claims. | Warranted caution versus failure to resolve conditions that available evidence supports. |

A model-only or direct-generation comparison is not automatically the right baseline for every question. A deterministic enumeration baseline, a fixed authored recipe, and a controller without one policy feature answer different questions. Later sections must identify which comparison isolates the claim being studied rather than treating all baselines as interchangeable.

### 3.7 Assessment independence and uncertainty

An assessment record should identify the claim examined, the reference criterion, the assessor/checker, the relevant versions, support and contradiction, and the resulting judgment. Automated checks can examine explicit constraints. Judgments involving the adequacy of natural-language refinements require separately stated reasoning and may remain unknown. Agreement between several AI agents does not transform an unsupported claim into proof.

| Assessment issue | Required interpretation |
| --- | --- |
| Multiple acceptable assemblies | Assess against requirements and permitted equivalences; do not demand one exact graph unless the case requires it. |
| Uncertain reference judgment | Preserve ambiguity; do not force a binary label solely to calculate a score. |
| Checker disagreement | Record the disagreement and adjudication grounds; do not hide it in a majority vote. |
| Repeated stochastic attempts | Retain proposal/output records and resource use; deterministic selection does not make the proposer deterministic. |
| Missing observations | Distinguish not attempted, interrupted, tool failure, blocked and algorithm-produced partial outcomes. |
| Statistical interpretation | Choose an appropriate analysis after defining sampling and dependence assumptions, before examining evaluation outcomes. |

No statistical observations exist for these hypotheses yet, so no empirical chart, effect size or significance claim is appropriate here. The tables describe the proposed logical structure of the study; they are not data about performance.

### 3.8 Decision consequences and limits on refutation

A single verified counterexample can refute an unconditional claim that an invariant always holds for an implementation. The empirical usefulness hypotheses are scoped predictions and need a declared comparison/decision criterion; one unfavorable case does not automatically refute all possible boundary-resolution methods. Conversely, successful examples cannot prove a universal synthesis claim.

| Observation category | Permitted next step | Claim to avoid |
| --- | --- | --- |
| Verified conformance violation | Diagnose code versus specification, preserve the failing artifact, and version the correction. | The failure is irrelevant because most cases passed. |
| Warranted residual on an unsupported case | Record support for that case-specific gap judgment. | The system discovers every missing requirement. |
| Conforming but unhelpful output | Reconsider policy, descriptions or the chosen usefulness criterion. | Structural validity alone demonstrates usefulness. |
| Better development result only | Retain as a development observation; examine fresh cases later. | Generalized improvement has been established. |
| Unclear or conflicting assessment | Record an inconclusive result and the missing adjudication basis. | Ambiguity is either a success or a failure by default. |
| Evidence supporting the initial transition | Consider a separately specified next-layer study. | Every higher plane follows automatically. |

### 3.9 Handoff to experimental design

Section 3 provides a hypothesis register and challenges to it. It does not authorize the paper to report findings. Before any evaluation, later sections must settle the exact input/output records, candidate-generation mechanism, case references, baseline versions, resource accounting, assessment procedures and decision criteria. Revisions to these commitments after observations must be disclosed with their timing and rationale.

The intended reasoning chain is: **claim → specified observation → recorded artifact and trace → independent assessment → scoped conclusion or explicit uncertainty**. Development can iterate, but evidence must remain attached to the version and conditions that produced it. This keeps the study open to discovering that the descriptions, the controller policy, or the initial scope need substantial revision.

## 5. Development Cases and Held-Out Evaluation

### 5.1 Purpose and current status

Development and evaluation answer different questions. Development cases expose missing rules, implementation defects and inadequate descriptions so that the procedure can be revised. Held-out evaluation cases examine an identified revision on cases that did not guide those revisions. Regression cases check whether earlier behavior deteriorates. The same case can change role over time, but its exposure history cannot be erased.

This section specifies a proposed case-construction and separation protocol. It does not create a completed benchmark, claim that an untouched evaluation collection already exists, or report any executed cases. All case families below concern the initial Components → Composite scope unless explicitly marked as an unsupported-profile challenge.

| Collection | Permitted use | Claim it can support |
| --- | --- | --- |
| Development | Inspect inputs, expected properties, traces and failures; revise rules or code. | Observations explaining development decisions. |
| Regression | Revisit previously exposed cases after revisions. | Preservation or loss of earlier assessed behavior. |
| Held-out evaluation | Assess a frozen version under a prespecified protocol without using those cases to develop that version. | Scoped evidence of transfer beyond the development cases. |
| Fresh follow-up evaluation | Assess a subsequent revision on a newly isolated collection. | Evidence about that later revision under a new evaluation epoch. |

A held-out label is a statement about access and use, not a guarantee of independence or representativeness. Renaming a familiar case or changing its wording does not make it fresh.

### 5.2 Case unit and case record

A case is an identified design problem together with separately maintained assessment information. It may specify a supported construction, a deliberately missing fact, a contradiction, or a condition whose correct resolution is genuinely uncertain. It need not prescribe one exact graph.

| Case-record field | Required content | Visibility |
| --- | --- | --- |
| case_ref / revision | Stable case identity and content revision. | Generator and assessor. |
| family_ref / ancestry | Structural family, source, transformations and sibling cases. | Registry; assessment visibility governed by protocol. |
| generator_input_ref | Section 4 input bundle: baseline, library, rules, profile, initial specification and policy. | Generator. |
| reference_properties | Required properties, acceptable alternatives and known unresolved conditions. | Assessor only unless explicitly supplied as an experimental condition. |
| justification | Grounds for reachability, contradiction, introduced gap or expected invariant. | Assessor. |
| exposure_record | Who or what accessed which parts, and whether access influenced a revision. | Study audit record. |
| collection / epoch | Development, regression or evaluation assignment and its effective version. | Study registry. |
| disposition | Retained, revised, excluded, exposed or quarantined, with reasons. | Study registry. |

The generator receives requirements and available descriptions, not a hidden answer key. A requirement's acceptance criterion may legitimately be visible; a complete reference solution or known failure location is additional information whose disclosure must be stated. Reference uncertainty remains a recorded property of the case.

### 5.3 Proposed case families

The families are chosen to exercise distinct obligations rather than merely vary the subject matter of a summarizer. They are design targets for later case authoring, not a finalized sample allocation.

| Family | Construction idea | Principal question | Reference basis needed |
| --- | --- | --- | --- |
| Supported assembly | A permitted composite can be formed from supplied descriptions without a complete recipe. | Can the procedure select, configure and connect an admissible arrangement? | At least one independently justified solution within the declared domain and horizon. |
| Relevant alternatives and distractors | Include plausible but unsuitable components alongside appropriate ones. | Does selection respect interfaces and requirements rather than names alone? | Reasons each candidate is acceptable, conditional or unsuitable. |
| Missing primitive | Omit a required realizable capability from the permitted component/rule domain. | Does the procedure report an unsupported leaf rather than invent an implementation? | The omitted capability and the declared limits supporting that gap judgment. |
| Missing information or binding | Leave a consequential parameter, intent choice or resource reference unresolved. | Does it distinguish clarification from a design reference and deferred activation? | Which fact is missing, why it matters and whether design can proceed conditionally. |
| Interface mismatch | Supply incompatible endpoints with no permitted adapter, or an explicit valid adapter alternative. | Are coercions rejected unless supported? | Endpoint declarations and permitted conversion rules. |
| Conflicting mandatory constraints | Introduce an explicit contradiction in a candidate or in a bounded case domain. | Does it retain the conflict without silently weakening the goal? | Scope of the contradiction; candidate-specific versus domain-wide grounds. |
| Nontrivial packaging boundary | Select members with crossing edges, exposed configuration and required unbound inputs. | Are boundary relations retained when represented above? | Independently enumerated membership and boundary incidence. |
| Shared identity and multiplicity | Use shared references and distinct occurrences of the same definition. | Are sharing and deliberate duplication distinguished? | Intended occurrence identities and membership relationships. |
| Changed evidence context | Change a relevant definition, configuration or dependency version. | Are affected judgments reassessed rather than inherited blindly? | Dependency links and the evidence's declared applicability. |
| Search limitations | Supply alternatives requiring branching or exhaust declared limits. | Are partial outcomes and omitted branches reported honestly? | Search registry, limits and retained trace; known solution only where justified. |
| Unsupported profile or cycle semantics | Request a move beyond the supplied profile or a feedback arrangement unsupported by the DAG profile. | Does the procedure return a profile/semantics gap? | Explicit limits of the active profile, without claiming the intent is universally impossible. |

A case may exercise several families. The registry should record primary and secondary mechanisms so that one complex example is not counted as several independent successes. Trivial cases are useful for diagnosis but cannot alone establish robustness on interacting constraints.

### 5.4 Construction sequence and admissibility of cases

| Preparation step | Required action | Guard against bias |
| --- | --- | --- |
| State the mechanism | Identify the rule, invariant or hypothesis the case examines. | Do not author a case merely because the current implementation already solves it. |
| Prepare the input | Pin requirements, descriptions, initial state, profile and policy assumptions. | Avoid hiding essential facts in the answer key that the generator was expected to know. |
| Establish reference grounds | Document a valid construction, a supported contradiction, or the exact introduced absence. | Do not use the evaluated controller's output as the sole reference. |
| Record permissible alternatives | State which distinct graphs or residual dispositions can satisfy the case. | Avoid exact-match scoring when multiple solutions are legitimate. |
| Assign family and ancestry | Link transformations and shared source examples. | Keep close variants from masquerading as independent problems. |
| Assign collection | Apply declared family-level separation and record access. | Do not move a difficult case out of evaluation after seeing failure. |
| Freeze the case version | Preserve its content and reference rationale before use. | Corrections create revisions with their timing disclosed. |

A “known solvable” label requires a construction admissible under the actual profile and available rule/operand domain. A possible runtime solution outside that domain does not establish controller reachability. Conversely, a finite search failure is insufficient to label a case unsatisfiable. Domain-wide impossibility should be asserted only when independently justified within a precisely stated finite domain; otherwise the reference remains limited or unknown.

### 5.5 Family-level separation and leakage

Development and evaluation must differ in the structure that the revision could have learned, not just names, identifiers or prose. Closely related cases belong to an ancestry group. Later sections must define the grouping and assignment method before evaluation; no randomization or sample size is selected here.

| Leakage route | Example | Required treatment |
| --- | --- | --- |
| Textual rewrite | A development case reappears with different wording. | Keep it in the same exposed family; do not count it as fresh transfer. |
| Graph renaming | Identical topology and decision with renamed nodes. | Record as a sibling/robustness variant rather than independent held-out evidence. |
| Reference leakage | The proposer receives a complete expected wiring or gap list. | Disclose the condition; do not claim discovery. |
| Developer inspection | A developer reads evaluation failures and adjusts a rule. | Mark those cases exposed for subsequent revisions. |
| Aggregate feedback | Repeated evaluation summaries guide policy choices. | Record the feedback exposure; fresh evaluation is needed for later claims. |
| Shared author/model familiarity | Case author or proposer may have seen similar public examples. | Disclose the risk; hidden local labels do not prove absence of prior familiarity. |

Family separation reduces obvious leakage but does not prove statistical independence. Multiple observations from one family, repeated model calls and paired variants must be analyzed with their dependence acknowledged. Any future statistical inference depends on the actual sampling protocol, not on the existence of an evaluation folder.

### 5.6 Exposure lifecycle and evaluation epochs

**Proposed lifecycle:** author and document → assign and freeze → develop on exposed cases → freeze the candidate procedure and assessment protocol → assess held-out cases → archive observations → mark evaluation feedback exposed if used for revision → acquire a fresh collection for a later evaluation epoch.

| Event | Case status after the event | Permitted interpretation |
| --- | --- | --- |
| Case used to fix a rule or prompt | Development-exposed; reusable for regression. | Evidence of development behavior, not untouched evaluation of the revised rule. |
| Frozen evaluation performed without adaptation | Evaluation observation for that frozen version. | Scoped assessment under the recorded conditions. |
| Evaluation output used to choose a revision | Exposed to the development process. | The earlier observation remains valid for the earlier frozen version; later claims need fresh assessment. |
| Case input or reference corrected | New case revision; original retained. | Explain whether earlier comparisons remain interpretable. |
| Suspected leak or invalid reference | Quarantined pending diagnosis. | Preserve the observation and reason; do not quietly delete it. |

This section does not promise technically enforced secrecy. The later implementation must specify storage/access controls or acknowledge an administrative separation. If one assistant has already inspected a proposed held-out case, that exposure must be recorded rather than denied.

### 5.7 Development cycle and regression set

Development may proceed slowly and interactively. Each iteration should identify a failure or missing rule, diagnose it, propose a specific change and preserve the before/after versions. A revised requirement, richer catalog description and altered search policy are different interventions and should not be bundled invisibly.

| Development record | Contents |
| --- | --- |
| Trigger | Case and artifact that exposed the issue. |
| Diagnosis | Implementation defect, inadequate description, rule/profile gap, policy limitation, requirement problem, or unresolved mixture. |
| Intervention | Exact code, rule, description, policy or baseline change. |
| Intended effect | The specific failure the change is meant to address. |
| Regression scope | Earlier exposed cases relevant to preservation and interaction effects. |
| Remaining uncertainty | Claims that the development comparison cannot establish. |

Regression is not a substitute for fresh evaluation. It is particularly useful for detecting identity loss, dropped obligations or new false gaps after a targeted fix. A repaired example should remain in the exposed collection even if its source text later changes.

### 5.8 Paired variants and sensitivity questions

Controlled variants can help isolate which information the procedure uses. They must alter a declared feature and retain other relevant context where possible. Expected relations are conditional on the stated profile; they are not universal semantic equivalences.

| Variant | Proposed comparison | Limitation |
| --- | --- | --- |
| Remove a necessary description/binding | Observe whether the previously supported condition becomes unresolved. | Removing irrelevant metadata need not affect the result; necessity must be justified. |
| Add an explicit compatible adapter | Examine whether a previously incompatible connection becomes representable. | Adapter behavior still needs its own contract/evidence. |
| Change a dependency version | Examine invalidation of affected judgments. | Only evidence linked to the changed context should be invalidated. |
| Add irrelevant alternatives | Examine selection and resource sensitivity. | Lexicographic ordering may legitimately change; identical output is not required. |
| Increase search allowance | Examine what additional alternatives become reachable. | More budget does not guarantee a better result under every policy. |
| Change layout only | Examine whether semantic assessment remains unchanged. | Stable identifiers and actual semantic records must remain fixed. |

These relations can later become checks or exploratory analyses after their exact conditions are specified. They are not executed tests in this section. Sibling variants remain grouped for interpretation and collection assignment.

### 5.9 Reference judgments and adjudication

| Reference type | Suitable grounds | Inadequate substitute |
| --- | --- | --- |
| Structural validity | Explicit endpoints, profile rules and independently inspectable mappings. | Visual appearance of connected nodes. |

Reference disputes should preserve both positions and the reasons for adjudication. If the reference cannot be settled, mark the affected criterion indeterminate. Do not force a favorable binary score. Internal AI review must be labeled as such and is not external academic peer review.

## 11. Discussion, Limitations, and Next-Layer Gate

### 11.1 What the code gives us

The notebook now performs real local structural experiments and produces inspectable JSON artifacts. It connects descriptions to a small selection procedure, packages selected occurrences, records guards and residuals, and exposes concrete failure modes. It is a useful executable starting point for the theory, not the completed general algorithm envisioned across the library stack.

| Limitation | Consequence | Next research need |
| --- | --- | --- |
| Synthetic descriptions and capability tags | No demonstration of real Langflow catalog adaptation or semantic understanding. | Pin and map real definitions, preserving unsupported metadata as unknown. |
| Source–transform–sink grammar | No arbitrary DAG synthesis, refinement bridge, recursive decomposition or frontier backtracking. | Implement and examine one additional rule family at a time. |
| Same-author cases and checker | Overfitting/reference mistakes remain plausible. | Freeze a revision before obtaining separately authored, access-controlled cases. |
| Weak baseline | Comparison cannot isolate all benefits of Paper 2. | Compare a matched exhaustive finite enumerator and targeted policy ablations. |
| Planned defect and repair | No evidence of autonomous diagnosis or independent revision discovery. | Record a genuine development cycle on new failure cases. |
| Fixed explicit constraints | Natural-language entailment, conflicting evidence and stale-judgment propagation are not implemented. | Add scoped judgment/evidence machinery before claiming those capabilities. |
| Partial requirement accounting | The prototype records fixed design/behavior/activation obligations rather than a general requirement ledger. | Implement full baseline/refinement/disposition records. |
| No generated behavior execution | Structural success cannot establish faithful summarization or operational readiness. | A separately scoped behavioral study if and when requested. |
| Notebook imports local study module | Reproduction requires this repository and recorded module version. | Preserve the module, protocol and environment with the notebook. |

### 11.2 Gate for extending upward

Before claiming Composite → Flows, the study needs an explicit flow profile for ordering, branches, outcomes and failure obligations; a versioned interface mapping; independently justified cases; and evidence that the implemented transition preserves its invariants. A complete lower-layer record alone does not define those semantics. It is appropriate to stop at this stage and report a missing-profile gap rather than fill higher planes with ungrounded labels.

Before broadening the present Components → Composite claim, priority work is a real component adapter, a general obligation ledger, complete boundary extraction in nontrivial ambient graphs, automatic cycle checks, and bounded alternate-branch search. Each addition needs its own case families and independent assessment. More planes in the diagram are not a substitute for these mechanisms.

### 11.3 Reproducibility and side effects

The executable notebook uses only local Python, explicit synthetic dictionaries and Matplotlib. It writes the study's observation files and displays tables/figures. It does not authenticate resources, call model endpoints, execute Langflow flows, modify reusable library definitions or deploy agents. The hashes in the manifest identify the protocol, module and case content used. They do not provide tamper-proof custody or independent validation. Re-running overwrites the current observation files; preserve a versioned copy before any later code/protocol revision intended for comparison.
