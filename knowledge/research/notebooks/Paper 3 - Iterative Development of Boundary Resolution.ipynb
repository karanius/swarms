### 2.4 Obligations inherited without relaxation

The study may revise a policy or propose a new theory version, but it must not silently relax the meaning of the baseline while reporting a comparison. The following obligations are inherited as conformance targets. They are not claims that the current code already satisfies them.

| Obligation | Consequence for an accepted design revision | Failure to record |
| --- | --- | --- |
| Requirement preservation | Retain every mandatory in-scope requirement and its judgment, or record an explicit baseline revision. | A hard requirement is deleted to manufacture completion. |
| Judgment integrity | Preserve met, violated and unknown distinctly; unresolved conflicting evidence remains outward unknown. | A model's assertion or user preference becomes behavioral proof. |
| Identity/version preservation | Retain occurrence identities and pinned definitions through packaging and unfolding. | A shared stateful member is silently cloned. |
| Complete boundary mapping | Account for crossing endpoints, required unbound inputs, exposed ports/parameters and dependencies. | A wrapper hides an unresolved input or dependency. |

| Evidence scope | Reassess affected judgments when configuration, definitions or dependencies change. | An altered assembly inherits unrelated evidence. |
| Trace continuity | Preserve reasons, rejected candidates, residual dispositions and limits across branches. | A failed path disappears from the account. |
| Resource-reference hygiene | Describe required authorized bindings without embedding secret values. | Credentials become part of a portable research artifact. |

### 2.5 Decision procedure inherited as a baseline proposal

| Decision stage | Inherited convention | What remains to investigate |
| --- | --- | --- |
| Extract boundary | Link an unresolved condition to its upper obligation and available or absent realization. | Whether the supplied descriptions expose the important mismatch. |
| Select boundary | Resolve prerequisites first; then use mandatory status, declared impact rank and stable order. | Whether that priority is useful on the chosen cases. |
| Generate proposals | Produce a finite, attributed batch from a declared registry/domain and caps. | Candidate coverage and the consequences of omissions. |
| Gate proposals | Reject violated mandatory guards; preserve unknowns according to structural/semantic class. | Correct implementation and adequacy of the guards. |
| Order candidates | Use the declared lexicographic policy, including conditional status and action rank. | Sensitivity to ordering; no optimality is assumed. |
| Apply and reassess | Accept one atomic design patch, retain residuals, invalidate affected judgments. | Whether recorded changes preserve the inherited obligations. |
| Return or branch | Use finite frontier/budget rules and explicit outcome records. | Usefulness of returned partial records and limits of the search horizon. |

Paper 2's action preference begins with specified reference resolution, reuse, composition and requested packaging, before supported decomposition and other declared revisions. Clarification is an information gate; it is not an arbitrary cheap candidate. This is an authored baseline convention, not an empirically discovered best policy. Deterministic, model-driven, neural and multi-agent proposal mechanisms remain possible alternatives; none is selected or validated by this section.

The finite-stopping argument is conditional: a finite charged-attempt budget bounds attempts only when each helper also returns or can be interrupted in finite time, and bookkeeping remains finite. It establishes neither solution discovery nor a practical latency bound. Iteration through controller revisions must also be distinguished from cycles inside an assembly graph. The initial conservative DAG profile does not provide semantics for arbitrary feedback systems.
