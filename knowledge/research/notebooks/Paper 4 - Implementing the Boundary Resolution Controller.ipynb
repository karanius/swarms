# Implementing the Boundary Resolution Controller — Paper 4

## Abstract

This study implements a bounded subset of Paper 2's boundary-resolution controller over an explicit typed structural-refinement grammar. Unlike Paper 3's fixed source–transform–sink pattern, the prototype can select direct components, expand declared refinements recursively, compose variable-length graphs, resolve described resource references, package the same occurrences, and revisit alternative branches. Three guarded policy variants are examined on four development and twenty-four same-author designated evaluation cases. Both retained-frontier variants match all twenty-eight references; the no-frontier policy matches twenty-six, missing two recoverable assemblies. No invalid accepted artifact is detected by the stated assessor. Priority scheduling uses 401 charged operations versus FIFO's 410 in the main suite, with identical reference outcomes. Seven deliberate corruptions are detected, and a separate scoped evidence helper preserves unknown judgments after context change or conflicting support. These are descriptive results on exposed synthetic cases, not independent validation. Structural bridges are supplied axioms; behavioral correctness, activation, all-layer generation and complete implementation of Paper 2 remain unestablished. Executable cells, traces, source freezes, JSON results and figures accompany the paper.


## 1. Research Question and Theory Traceability

**Question:** Can an explicit boundary-resolution controller repeatedly choose a permitted action, revise an instance specification and preserve its obligations until it either constructs a composite or returns a justified stopping disposition?

Paper 2 is the source of the controller design. Paper 3 demonstrated a narrower finite pipeline selector and packaging correction. Paper 4 advances the controller implementation, rather than replacing the theoretical direction with an unrelated generator. Its “discovery” is selection and recursive expansion within a supplied finite grammar, not invention of new capabilities or unconstrained reasoning about text.

| Paper 2 mechanism | Implementation in this study | Limit |
| --- | --- | --- |
| Pinned β, L and π | Requirement/library versions plus identified profile, policy and implementation. | Versions are local identifiers with frozen artifact hashes; no live catalog refresh. |
| Working Q | Occurrences, named edges, resource bindings, local representations and refinement records. | Components → Composite only. |
| J and U | Per-obligation judgments, root accounting, dependencies and unresolved residuals. | Structural task DSL, not unrestricted predicates. |
| Boundary extraction/selection | Open obligations become boundaries; ready prerequisites gate selection; priority uses mandatory, impact, order and identity. | No inference of unstated user requirements. |
| Candidate A and guard gate | Finite direct-definition/refinement candidates; nominal types, locality, bridge and cycle guards. | Supplied symbolic capabilities and structural axioms. |
| Alternative F | Bounded branch snapshots; restoration retains global trace, visited set and budget. | No completeness under pruning. |
| Visited V | Canonical semantic state digest excluding display coordinates and trace timestamps. | No graph-isomorphism or semantic-equivalence solver. |
| Trace T and budget b | Charged propose, assess, apply and restore events with rejected alternatives and before/after digests. | Memory/file history, not tamper-proof custody. |
| Evidence invalidation | Separate helper checks version context and conflicting support. | Helper examination only; dynamic controller epochs are not implemented. |

### 1.1 Claims examined

| Claim | Proposed examination | Refuting or limiting observation |
| --- | --- | --- |
| C1 — repeated controller transitions | Direct, two/three-step and nested refinement cases. | Unsupported closure, lost requirements, invalid mappings or exhausted known route. |
| C2 — alternative-branch value | Compare retained-frontier policies to the same priority policy without alternatives. | No recoverable difference, or unsafe recovery. |
| C3 — scheduling effect | Compare priority/depth-first to ready insertion-order/FIFO with common guards/domain/budget. | Same or worse operation counts; no general speed claim from one suite. |
| C4 — honest boundaries and stops | Missing resources/primitives, ambiguity, cycles, pruning and budgets. | Fabricated success, dropped obligations or impossibility claims from finite failure. |

The comparisons are policy bundles: FIFO changes both ready-boundary order and frontier retrieval order. They do not isolate each policy dimension causally. A no-frontier failure may be safe and honest while still missing an available solution; safety and solution coverage are separate measures.


## 2. Implemented State and Rule Grammar

### 2.1 Case and controller records

| Record | Contents | Preservation obligation |
| --- | --- | --- |
| Requirement | ID, explicit capability, input/output type, locality, mandatory status and prerequisite IDs. | Every root persists in U/J with behavioral judgment unknown. |
| Definition | Pinned identity, Components layer, named inputs/outputs, configuration and resource references. | Occurrences retain original description fields. |
| Refinement | Capability signature, named child tasks, directed child connections, entry/exit and bridge judgment. | Children plus structural bridge remain represented. |
| Obligation | Identity/owner, kind, readiness dependencies, depth, priority and status. | Root status changes only through accepted realization/composition actions. |
| Candidate | Rule/definition identity, action, guards and deterministic ordering key. | Unknown/violated structural bridge cannot be committed. |
| Outcome | Scoped disposition, residuals, attempts, frontier omissions and full current state. | Blocked within horizon is never a universal impossibility assertion. |

A requirement supplies a **structural** input/output signature. A refinement's bridge is a supplied structural axiom: the named child graph represents that capability under the study grammar. It does not prove behavioral implication. Direct reuse is considered before decomposition following Paper 2's action preference. Recursive refinements stop at the declared depth limit.

### 2.2 Actions and invariants

| Action | Effect | Check |
| --- | --- | --- |
| Select/reuse | Create a stable task-derived occurrence referencing a pinned definition. | Nominal input/output, layer and locality guards. |
| Decompose | Replace the task's pending realization with child obligations and a join. | Explicit matching signature, nonempty refinement, met bridge and acyclic child relation. |
| Resolve reference | Record a supplied described resource binding. | Does not authenticate or assert activation. |
| Compose | Join completed children using declared named edges or complete a bound leaf. | Endpoint/type legality and component-graph cycle checks before commitment. |
| Package | Write one top representation plus local nested representations referencing the same members. | Required unbound inputs, crossings, config and dependencies exposed. |
| Backtrack | Restore an alternative branch snapshot. | Global budget, trace and visited history are retained. |
| Clarify/return | Ask a capability question or report conflict/gap/limit. | No invented missing implementation or higher profile. |

Root prerequisite cycles and proposed refinement cycles are detected from actual relations; this improves on Paper 3's explicit cycle flag. Configuration/resource descriptions are retained, but stateful runtime scheduling and concurrency are unmodeled. Two root requirements can create distinct occurrences of the same definition. Nested composites are local representations of lower members, not an implemented Composite → Flow transition.

### 2.3 Bounds and charging

| Limit | Default | Meaning |
| --- | ---: | --- |
| Charged operations | 160 | Positive unit for propose, candidate assessment, accepted application and frontier restoration. |
| Frontier capacity | 16 | Alternate snapshots retained; omissions explicitly counted. |
| Refinement depth | 5 | Maximum expansion depth. |
| Node count | 24 | Accepted state-size guard. |
| Candidate batch | 8 | Maximum candidates assessed at a selected boundary. |

Some cases override these values. Helpers operate synchronously over small authored local data. This is not hardened execution of arbitrary untrusted rule services; no wall-clock guarantee for unbounded external helpers is established. Termination arguments concern finite attempts under those local-helper assumptions, not solution discovery. Candidate ordering uses the supported projection of Paper 2's lexicographic convention; it does not implement every proposed ranking field or optimize expected reward.


## 3. Protocol, Cases, and Comparators

### 3.1 Collection and disclosure

The suite contains four development cases and twenty-four designated evaluation cases, all authored and visible to the same assistant. It is **not independently held out**. The oracle is removed from the input passed to the controller, but that separation does not remove author familiarity or guarantee a correct reference.

| Case group | Mechanisms examined |
| --- | --- |
| Construction | Direct reuse, two/three-step expansion, nested refinement, multiple roots and irrelevant alternatives. |
| Alternatives | A first refinement has an unsupported leaf; a later declared refinement can succeed. |
| Guards | Type/locality rejection, unknown bridge, incompatible/no component. |
| Dependency handling | Described/absent resources, root prerequisite order, missing prerequisite, prerequisite cycle. |
| Packaging | Extra required input, ambient crossing edge, same-definition distinct occurrences. |
| Stops | Ambiguity, mandatory conflict, empty scope, depth/budget/frontier limits and unsupported profile. |

Case references specify expected disposition/gap prefix and, where appropriate, structural counts, boundary properties and a backtracking witness. The assessor independently inspects accepted artifacts against the pinned definitions and refinement graph. It shares authorship with the generator and is not external review.

### 3.2 Methods and decision criteria

| Method | Boundary scheduler | Alternative policy | Shared safeguards |
| --- | --- | --- | --- |
| boundary | Ready prerequisites, then Paper 2-inspired priority. | Depth-first retained frontier. | Typed guards, structural invariants, limits and scoped outcomes. |
| fifo | Ready prerequisites, then insertion order. | FIFO retained frontier. | Same grammar and safeguards. |
| greedy | Same priority as boundary. | No retained alternatives. | Same grammar and safeguards. |

| Measure | Definition | Interpretation |
| --- | --- | --- |
| Case pass | Expected outcome/gap plus reference artifact checks. | Case-specific conformance/coverage, not population accuracy. |
| Invalid accepted | A design-closed artifact fails reference checks. | A safety failure, separate from honest blocked outcomes. |
| Charged operations | All proposal, assessment, application and restoration units. | Defined bookkeeping; not elapsed latency or model cost. |
| Backtracking witness | Trace contains retained-state restoration on designated recovery cases. | Demonstrates this recovery mechanism, not complete search. |
| Negative checks | Deliberately corrupt accepted results and inspect assessor rejection. | Checks sensitivity to those faults, not assessor completeness. |

A boundary-method acceptance target is all authored suite references matched and zero invalid accepted artifacts. All comparator failures remain reported. No statistical test, significance threshold or generalized reliability estimate is appropriate for this small exposed collection.

## 6. Discussion and Limits

### 6.1 What changed from Paper 3

Paper 3 selected a transformation inside a supplied pipeline. Paper 4 implements a repeated decision process over obligations: select a ready boundary, obtain a bounded batch, check guards, accept one design revision, retain alternatives, reassess readiness and return a scoped outcome. This supplies executable instances of several Paper 2 mechanisms. It still depends on a declared library and refinement grammar; the algorithm cannot invent missing semantic knowledge merely because its control loop is more general.

| Paper 3 boundary | Paper 4 advance | Remaining limitation |
| --- | --- | --- |
| Fixed source/worker/output shape | Direct, variable-length and nested declared refinements. | Arbitrary topology generation and new rule discovery are absent. |
| Simple candidate scan | Repeated ready-boundary selection and guarded transitions. | Input predicates are a small structural DSL. |
| No alternate branch restoration | Bounded snapshots with global budget/history retained. | No completeness after batch/frontier pruning. |
| Limited obligation record | Root/child/binding/join/package obligations and unknown behavioral judgments. | Not a general AND/OR logical entailment engine. |
| Cycle flag | Actual prerequisite and declared refinement cycle checks plus accepted-graph DAG check. | Supported cyclic runtime semantics are not modeled. |
| No context invalidation code | Scoped evidence helper and explicit unknown conflict results. | Dynamic controller epochs remain unimplemented. |

### 6.2 Where Paper 2 remains ahead of the implementation

| Unimplemented or partial mechanism | Why it matters | Required next work |
| --- | --- | --- |
| Conversational specification | The controller assumes explicit confirmed structural inputs. | A separately assessed interviewing/normalization bridge. |
| General requirement refinement | Bridges are supplied structural axioms. | Richer criteria and independently justified refinement relations. |
| Candidate-method comparison | Only deterministic explicit proposals are used. | Fair deterministic/model/Prolog/neural proposal comparisons under shared guards. |
| Full policy key | Supported fields project Paper 2's ordering. | Exact field semantics and targeted ablations before claims about the whole policy. |
| Full dynamic epoch/reassessment | Helper outputs are not integrated with changing library context in the main loop. | Dependency indexing, stale-judgment propagation and explicit epoch transitions. |
| Progress vector | Trace is recorded, but the full proposed change vector is not computed. | Derived obligation/structural/ambiguity changes per accepted patch. |
| General shared ownership | Task-derived occurrence IDs preserve this grammar's identities. | Overlapping wrappers, resource sharing and stateful concurrency semantics. |
| All-layer profiles | Only Components → Composite semantics are implemented. | Separately define Flow/Process profiles rather than extrapolate numeric levels. |
| Robust external helpers | Local authored data and finite Python helpers only. | Interruption, resource isolation and malicious/unbounded input analysis if externally supplied rules are allowed. |

A missing primitive remains unsupported. A local composite representation is not automatically registered into the library. Resource binding records remain references; no credentials are embedded and no authentication is performed. Behavior and activation do not become met when design closes.

### 6.3 Threats to validity

The author supplied code, grammar, cases, reference expectations and interpretation. The reference assessor uses separate code and the controller never receives oracle fields, but those facts do not make the study independently reviewed. A deterministic implementation can still share a misconception with its checker. Fault injection establishes detection of seven selected corruptions, not proof that every error is detectable.

The collection is small, deliberately structured and partly includes related development/evaluation mechanisms. A backtracking benefit on two sibling examples is not evidence of prevalence across arbitrary tasks. FIFO is a meaningful guarded comparison, but neither it nor the boundary policy is an exhaustive independent planner over all possible graphs. The greedy comparator tests retention of alternatives; its fewer operations partly result from stopping unsuccessfully.

No confidence intervals, population success rates or significance tests are reported. No runtime task behavior, live component interoperability, latency, cost, user comprehension or deployment readiness was measured. The finite attempt bound assumes bounded local candidate construction/application; it does not prove the termination of arbitrary future model or tool services.

### 6.4 Next research gate

The next defensible step is to freeze this revision and have a separate author create genuinely unseen structural cases, including nontrivial branching/fan-in refinements and ambient boundaries, plus a pinned real Langflow definition adapter. Only then should broader solution-coverage claims be considered. In parallel, dynamic evidence epochs and the full obligation model can be implemented as explicit new mechanisms with their own tests. Extending to higher planes requires supplied versioned profiles and new acceptance conditions; the present success does not license arbitrary Process or Enterprise generation.


## 7. Conclusion, Review, and Evidence

### 7.1 Conclusion

Paper 4 delivers an executable, bounded controller subset of Paper 2. It repeatedly resolves explicit structural obligations, expands declared refinements, builds and packages a composite graph, and either closes the design or preserves reasons for clarification, blocking or exhaustion. In the final suite, both frontier-based variants match all 28 authored references; no-frontier selection matches 26. The difference demonstrates a concrete value of retained alternatives in the examined grammar. Priority scheduling changes charged work but not solution coverage in this suite.

This is progress from a fixed structural recipe toward an inspectable rule controller, not the completion of general instance generation. The strongest remaining uncertainty is transfer to independently authored problems and real component metadata. Full semantic refinement, dynamic epochs, behavioral validation and higher-plane profiles remain explicit gaps.
