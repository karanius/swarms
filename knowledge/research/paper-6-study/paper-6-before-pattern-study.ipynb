# Paper 6 — Rule Mining from Langflow Templates

## Abstract

This study will investigate whether relationships in a provenance-preserving corpus of Langflow templates can support explicit, testable candidate rules for composite formation. Source graphs remain examples rather than validated composites. Configuration dependencies, native component identities, typed connections and packaging boundaries define the initial analytical scope. Mining methods, evaluation and findings remain to be developed; collection alone establishes neither rule validity nor behavioral correctness.

The fourteen sections organize the manuscript; the five phases below organize its development. Writing, implementation decisions, failures and evidence will be recorded throughout. These are planned phases, not completed results.

**First milestone:** One explicit mined rule with an inspectable antecedent and consequent, supporting examples, observed exceptions or an explicit statement that none were observed, defined support and confidence denominators, and links to original sources. This establishes an inspectable candidate, not behavioral validity or general composite formation.

**Iteration:** Findings may send us back to an earlier phase. Preserve the original protocol and version each revision; keep development feedback separate from evaluation evidence. Expand only after the first rule and its evidence can be checked.


### Updated Method Strategy — Association Discovery Before Relational Mining

**Decision:** Introduce association rule mining as an exploratory first stage. Retain Python/Jupyter, PyClause and AnyBURL for the subsequent relational stage. Prolog remains a candidate for richer explicit reasoning; no implementation winner or speedup has been measured. This decision supersedes the earlier recommendation to begin directly with relational mining.

| Stage | Proposed tool / artifact | Question answered | Handoff |
|---|---|---|---|
| 1. Freeze and normalize | Pinned corpus, provenance and typed extraction | What is actually present and observable? | Source-linked graph facts and eligible transactions |
| 2. Association discovery | mlxtend FP-Growth followed by association_rules | Which declared features co-occur, and which associations exceed their background prevalence? | Ranked exploratory associations with support, confidence, lift, exceptions and family coverage |
| 3. Inspect and design the relational encoding | Mapping ledger and source witnesses | Which associations suggest meaningful predicates, context or search restrictions? | Documented encoding choices; retained full eligible facts and a baseline encoding |
| 4. Relational rule learning | PyClause with AnyBURL; AMIE as a possible comparison | Which explicit variable-linked relationships can be learned in the chosen representation? | Candidate relational rules, exact evidence accounting and declared engine scores |
| 5. Independent checks and interpretation | Separate-family evaluation and boundary checks | Do rules transfer, and what do they actually establish? | Supported, challenged or inconclusive claims; no automatic composite promotion |

**Why the preliminary stage helps:** It can identify promising component combinations, typed edge features and resource dependencies to investigate. It cannot decide the correct ontology, recover ordering discarded by a basket representation, establish causal necessity, or prove a useful composite boundary. Speed and usefulness on this corpus remain hypotheses.

#### Transaction design comes first

| Exploratory view | Transaction / possible items | What must remain explicit |
|---|---|---|
| Whole graph presence | One eligible deduplicated graph; native component types and observable resource kinds | Presence ignores multiplicity, role and wiring; never interpret co-presence as a connection |
| Typed connection features | One graph; items encoding observed source type/port → target type/port relations | Direction and port identity must be encoded in the item itself; association does not add those semantics |
| Local neighborhood | One scoped occurrence neighborhood under a fixed definition | Many neighborhoods in one graph are dependent; use separate denominators and family-aware summaries |

Run these as separate analyses rather than mixing transactions with incompatible meanings. Do not include source IDs or unique occurrence IDs as predictive basket items. Keep them as provenance. Exclude configuration values, credentials and copied code from the item vocabulary. Distinguish observed absence from unknown extraction; record coverage and avoid silently replacing unknowns with False.

#### Critical controls

- Establish source families and the development/evaluation split **before** exploratory association mining. Choosing features using every graph would leak evaluation information.
- Count exact duplicates once while preserving every origin. Report both graph-level frequency and source-family breadth so mirrors cannot masquerade as independent support.
- Define the transaction population, minimum support/count, itemset-size cap, rule filters and parameter search bounds before runs. Candidate rule counts can grow combinatorially; runtime is not guaranteed by using FP-Growth.
- Report support and confidence together with lift and raw counts. High confidence can result from a ubiquitous consequent; high lift can rest on very few cases. Thresholds and uncertainty treatment remain to be specified.
- Associations suggest an encoding; they must not become the only facts supplied to the relational learner. Preserve the full eligible fact store and compare an association-guided encoding/search bias with a baseline that does not use that guidance.
- PyClause's AnyBURL and AMIE interfaces have their own rule languages and confidence definitions. Do not equate itemset confidence with their scores, or transfer a measured association into a hard constraint.
- Graph-local occurrence identities must prevent cross-template joins. Packaging, required-input preservation and behavioral validity remain separate checks.

**Immediate milestone:** An association evidence table linked to real source graphs, followed by a reasoned mapping from selected associations into a candidate relational representation. This is the new first experiment; no mining has run yet.

## Main finding

Use one provenance-preserving, typed relational representation, then expose separate views for motif discovery, explicit rule mining, boundary proposals and possible evolutionary search. These are different tasks. A recurring subgraph is evidence of recurrence; a conditional rule adds a stated antecedent, consequent and denominator; a reusable composite additionally needs an interface, purpose and obligations. None establishes behavioral correctness by itself.

**Recommendation, not a literature theorem:** begin with bounded explicit relational rules and exact evidence accounting. Use graph motifs as a comparison and candidate source. Treat typed genetic programming or graph evolution as a subsequent candidate-search comparison once the representation, legality checks and scoring protocol exist. NEAT is relevant inspiration for preserving innovation and exploring topology; it is not an existing Langflow-composition algorithm.

## Proposed common representation

This is a design synthesis informed by S1–S2 and S5–S12, not a schema provided by any single source. Keep JSON as interchange; derive mining-specific views from it without changing native components.

| Record | Minimum content | Why it is separate |
|---|---|---|
| SourceArtifact | original URL, immutable revision or explicit unknown, path, hash, collection time, license reference, source family | Many origins can share identical bytes; provenance remains many-to-one. |
| GraphOccurrence | artifact reference, nested graph path, native schema/version evidence, graph ID | One export can hold more than one graph. |
| ComponentOccurrence | graph-local ID, native definition/type/version reference, role label separately | Two Agent occurrences remain two occurrences of one type. |
| PortOccurrence | owner, direction, stable handle, declared accepted/emitted types, requiredness, selected output, dynamic-context evidence | A component-type pair does not uniquely identify a legal connection. |
| RelationFact | predicate, typed arguments, graph scope, evidence location, extraction version, observed/derived/unknown judgment | Facts may be reified for provenance; mining views can flatten them under declared mappings. |
| ConfigurationDependency | component/port reference, resource kind, required setting, external binding reference, missingness | Never store credential values in mining facts. |
| BoundaryProposal | members, internal edges, crossing-edge map, exposed required ports, scope/purpose hypothesis, unresolved obligations | A motif lacks these interface and obligation decisions. |
| CandidateRule | rule ID, typed variables, antecedent atoms, consequent, applicability conditions, unknown/negative policy, search bounds | Conditional structure must be inspectable, not only a name or score. |
| RuleEvidence | frozen corpus, graph/family/grounding witnesses, counterexamples, unknown cases, support/confidence definitions, selection history | Scores must be reproducible and must not erase exceptions. |
| EvolutionCandidate | rule or graph genome, operators, parentage, generation/seed, hard-constraint report, score vector | A search proposal is not a new native component or validated library item. |

Useful relation vocabulary to specify: `instanceOf`, `ownsPort`, `acceptsType`, `emitsType`, `connects`, `requiresResource`, `memberOf`, `crossesBoundary`, and `derivedFrom`. Every fact is scoped to a graph/source; unrestricted joins across unrelated templates would manufacture assemblies that were never observed. Port attributes and n-ary facts can be represented as explicit entities/records and projected to binary views when required by a miner.

### Example rule form (symbolic only)

`instanceOf(x, A) ∧ ownsPort(x, p) ∧ emitsType(p, T) ∧ instanceOf(y, B) ∧ ownsPort(y, q) ∧ acceptsType(q, T) ⇒ connects(p, q)`

This asks whether compatible typed A/B occurrences are connected under additional declared conditions. It does **not** say all compatible occurrences should connect. Define the opportunity population first: which x/y pairs count, whether required/optional ports differ, and which absent edges can be judged. Otherwise confidence can be meaningless. Source type names in a real rule come from the pinned corpus, not invented primitives.

### Evidence definitions to settle before coding

| Measure | Proposed definition / condition | Important limit |
|---|---|---|
| Graph support | Number of distinct eligible normalized graphs with at least one witnessed body-and-head grounding | Counts a graph once regardless of repeated occurrences. |
| Family support | Number of source families with witnessed body-and-head grounding | Requires explicit, audited family assignment. |
| Grounding confidence | Supported body-and-head groundings divided by assessable body groundings | Must define grounding identity, opportunity set and head completeness. |
| Unknown coverage | Body groundings whose head cannot be assessed, divided by all considered body groundings | Do not quietly remove unknowns and imply universal applicability. |
| Counterexamples | Assessable body groundings with explicitly contradicted or completeness-qualified absent head | Absence under an open-world policy is not a counterexample. |
| Boundary conformance | Separately checked crossing-edge and required-port preservation | Static conformance does not measure useful behavior. |

Graph-level support and AMIE head-prediction support are not interchangeable. Implementations must name the convention actually used. Family-aware numbers reduce mirror dominance but do not prove causal independence.



## Bounded pipeline and evolutionary branch

```mermaid
flowchart TD
 A[Freeze originals and provenance] --> B[Classify formats and versions]
 B --> C[Deduplicate bytes and group source families]
 C --> D[Extract scoped typed facts]
 D --> E[Validate extraction against source witnesses]
 E --> F[Bounded explicit rule mining]
 E --> G[Frequent-motif baseline]
 E --> H[Type-only candidate baseline]
 F --> I[Recompute witnesses, exceptions and unknown coverage]
 G --> I
 H --> I
 I --> J[Evaluate on separate source families]
 J --> K[Inspect candidate rules and boundary proposals]
 K --> L[Record findings, objections and revisions]
 E -. later controlled comparison .-> M[Typed GP or graph-evolution proposals]
 M --> I
```

Hard legality constraints should be checked separately from soft utility scores. Proposed evolution operators include type-preserving edge changes, bounded component insertion from the pinned library, and subgraph replacement preserving the declared interface. Each is a hypothesis about a useful search neighborhood, not a fact established by S10–S12. Required inputs, resource bindings and boundary mappings must survive or become explicit gaps. Cycles should follow a declared representation profile rather than assuming every archived Langflow export is a DAG.

For a fair later comparison, give bounded enumeration, motif-guided proposals and evolution the same eligible library, constraints, evaluation population and compute budget. Report search failures and variance across seeds. A fitness based only on corpus frequency will optimize resemblance, not task success. Introduce purpose/behavior objectives only when independent reference judgments or executable tests actually exist.

# Supplement — Concrete Representation and Evidence Decisions

# Paper 6 — Research-to-design decisions

Status: proposed design following a targeted primary-source review. No mining or evolutionary experiments performed. This is not an exhaustive or systematic review.

## Shared representation

Proposal: preserve the vendor graph export unchanged, then derive a versioned, typed, attributed, directed multigraph with explicit ports and an open boundary. Keep each source occurrence and source-family membership. A composite candidate references a set of these same occurrences and its exposed boundary; it is not inferred merely from a dense cluster.

[Fong's original thesis](https://arxiv.org/abs/1609.05382) formalizes networks of typed terminals and compositional boundaries. It motivates the interface discipline, but does not establish that Langflow's types instantiate that formalism or that our proposed composites behave correctly. [Speranzon, Spivak and Varadarajan](https://arxiv.org/abs/1802.03080) separately treat abstraction and behavioral contracts; their semantics cannot be inherited just by copying wiring.

| Record | Proposed fields | Purpose |
|---|---|---|
| SourceArtifact | source URL/path, revision, hash, collection time, license, format, fixture flag | Preserve exact evidence and eligibility |
| ComponentOccurrence | source graph, occurrence ID, native type/version, configuration references | Distinguish native definitions from occurrences |
| Port | occurrence, port ID, direction, declared types, required/unknown state | Preserve real interface constraints |
| Edge | source/target occurrences and ports, relation kind, source pointer | Separate data, tool and configuration relations |
| OpenBoundary | membership, incoming/outgoing edges, required unbound inputs | Avoid dropping obligations when proposing packaging |
| CandidateRule | antecedent, consequent, variables, guards, scope, counting semantics, witnesses, counterexamples, unknowns, version | Make implications explicit and auditable |
| EvidenceJudgment | claim, evidence kind, supported/refuted/unknown, provenance | Separate observed structure from inferred purpose and behavior |

## Rule language and evidence

Proposal: start with positive, range-restricted relational rules. Every consequent variable must be bound by the antecedent. Treat absence under incomplete extraction as unknown, not a negative fact. Treat packaging as a separate proposal needing declared criteria. Do not turn a frequent graph motif directly into a semantic implication.

For a fixed rule and declared unit of counting, define antecedent witnesses A and witnesses also satisfying the consequent J. An empirical closed-observation confidence would be |J| / |A| when |A| > 0. It is undefined at zero. This ratio is meaningful only if the consequent is observable for the eligible antecedent cases. Report unknown consequent status separately; do not silently remove unknown cases and present the result as unconditional confidence. AMIE-style PCA confidence, if chosen, must be implemented and named according to its actual definition rather than relabeling this ratio.

Report both artifact-level and source-family-level counts. Exact hashes remove byte-identical copies; they do not remove all tutorial, fork or common-ancestor dependence. Record raw counts, unique hashes and source-family coverage side by side. Thresholds and family definitions remain to be fixed before evaluation.

## Mining and evolution have different jobs

| Approach | Candidate object | Role in the proposed study | Necessary safeguard |
|---|---|---|---|
| Frequent-subgraph discovery | Repeated labeled subgraph | Find reusable structural motifs | No claim of implication or useful boundary from frequency alone |
| Relational rule mining | Conditional relationship | Produce explicit candidate rules with witnesses | Define missingness, denominators, negatives and dependence |
| Typed genetic programming | Typed program/rule expression | Optional search over rule bodies or candidate constructors | Grammar/type constraints plus independent evidence evaluator |
| Graph genetic programming | Indexed graph genotype | Optional search over proposed wiring | Preserve ports, required inputs and acyclicity only where required |
| NEAT | Neural topology and weights | Possible later learned proposal/ranking policy | A separate policy encoding and valid reward; not direct evidence for workflow synthesis |

[DEAP's official GP tutorial](https://deap.readthedocs.io/en/stable/tutorials/advanced/gp.html) documents typed primitives and warns that generation can fail when suitable terminals are unavailable. Typed generation is therefore a constraint mechanism, not a completion guarantee. [The original NEAT paper](https://nn.cs.utexas.edu/downloads/papers/stanley.ec02.pdf) concerns neural topology and weight evolution. Applying it to this corpus would require a new mapping and evaluation problem; rapid discovery of layer rules is unproven.

Recommendation: use one relationship representation for mining, deterministic checks and optional evolutionary proposals. Keep the checker outside the proposal process. Begin with relational mining and transparent baselines; add evolutionary search only as an explicitly measured comparison, not a presumed improvement.

## Evaluation and claim boundaries

[Process discovery](https://processmining.org/process-discovery.html) uses event logs; static workflow exports are not those logs. We can investigate design regularities now, not observed runtime conformance or performance. [Dwork et al.](https://papers.neurips.cc/paper_files/paper/2015/hash/bad5f33780c42f2588878a9d07405083-Abstract.html) show why adaptively reused evaluation data can overfit. Our practical proposal is a versioned exposure ledger and source-family-disjoint evaluation. We do not claim their formal guarantees without implementing their mechanisms and meeting their assumptions.

| Gate | Required evidence | What it does not establish |
|---|---|---|
| Corpus gate | All origins recorded; incompatible formats separated; snapshot frozen | Production quality |
| Extraction gate | Auditable examples and reference facts, including unknowns and boundaries | Runtime behavior |
| Rule gate | Explicit conditions, denominators, witnesses and exceptions | Causality or semantic necessity |
| Generalization gate | Predeclared evaluation on separate source families; revisions logged | All future libraries |
| Composite gate | Typed boundary and obligation preservation with justified formation criteria | Task success without behavioral tests |

## First bounded implementation, following the recorded protocol

1. Freeze eligible sources and assign source-family groups.
2. Extract occurrences, typed ports and actual edge relations with source pointers.
3. Hand-check representative exports and difficult cases before bulk mining.
4. Compare frequency-only motifs with bounded relational rules.
5. Produce one auditable rule card: meaning, variables, scope, counts, missingness, sources, exceptions and limitations.
6. Evaluate frozen rules on separate families. A failure revises the versioned design rather than rewriting the original hypothesis.

No experiment results are asserted here. The immediate deliverable is a literature-grounded representation and protocol; composite generation remains a later inference and validation problem.
